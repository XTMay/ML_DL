## 分词与词性标注

**目标：** 学习如何将文本分割成词语（分词），并识别每个词语的词性。


在语言学中，词语是最小的、可以独立运用的语言单位。将一段文本分割成一个个独立的词语是很多后续分析的基础，这个过程称为“分词”（Tokenization）。识别每个词语的词性（如名词、动词、形容词等）则称为“词性标注”（Part-of-Speech Tagging, POS Tagging）。

我们将使用一个非常流行的Python库 `NLTK` (Natural Language Toolkit) 来完成这些任务。NLTK是专门为自然语言处理设计的库，包含了许多常用的工具和语料库。

**Python操作：** 使用 NLTK 库进行分词和词性标注。

**语言学意义：** 分词对应于我们对文本进行“词语切分”，词性标注对应于我们对每个词语进行“语法范畴归类”。这是进行句法分析、语义分析等更高级研究的基础。

## 词干提取与词形还原

**目标：** 学习如何将词语还原到其基本形式（词干或原型）。


在语言中，同一个词语可能有不同的形态变化（如单复数、不同时态的动词等），但它们的核心意义是相关的。词干提取（Stemming）和词形还原（Lemmatization）是将词语还原到其基本形式的过程。

*   **词干提取**：简单地移除词语的后缀，得到一个词语的“词干”，词干可能不是一个实际存在的词。例如，"running" -> "runn"。
*   **词形还原**：将词语还原到其字典中的基本形式（原型），原型通常是一个实际存在的词。例如，"running" -> "run"。

词形还原通常比词干提取更准确，因为它考虑了词语的词性和上下文。

**Python操作：** 使用 NLTK 库进行词干提取和词形还原。

**语言学意义：** 这两种技术帮助我们将不同形态的词语归一化，便于进行词汇统计和基于词根的分析。这就像是识别出不同“形态变体”背后的“核心词汇”。

```python
from nltk.stem import PorterStemmer, WordNetLemmatizer
# 需要下载 WordNet 语料库用于词形还原
nltk.download('wordnet')
nltk.download('omw-1.4') # open multilingual wordnet
```

```python
stemmer = PorterStemmer()
lemmatizer = WordNetLemmatizer()

words = ["running", "runs", "ran", "beautiful", "quickly", "languages"]

print("原始词语:", words)
```

```python
# 词干提取
stemmed_words = [stemmer.stem(word) for word in words]
print("词干提取结果:", stemmed_words)
```

```python
# 词形还原 (需要指定词性，这里假设是动词或名词)
# 'v' 表示动词 (verb), 'n' 表示名词 (noun)
lemmatized_words_v = [lemmatizer.lemmatize(word, pos='v') for word in words]
print("词形还原 (动词):", lemmatized_words_v)
```

```python
lemmatized_words_n = [lemmatizer.lemmatize(word, pos='n') for word in words]
print("词形还原 (名词):", lemmatized_words_n)
```

## 特征工程

**目标：** 理解如何将文本数据转换为计算机可以理解的数值形式。

计算机无法直接理解文本，我们需要将文本转换为数值形式，这个过程称为“特征工程”。最简单也最常用的特征表示方法是“词袋模型”（Bag-of-Words）。

**词袋模型：** 将文本看作是一个装满了词语的“袋子”，忽略词语的顺序和语法结构，只关注每个词语出现的频率。我们可以创建一个词汇表，然后统计每个文本中每个词语出现的次数，得到一个向量来表示文本。

**Python操作：** 使用 scikit-learn 库计算词语频率。

**语言学意义：** 词袋模型将复杂的语言结构简化为词语的“出现频次”信息。虽然损失了语法和顺序信息，但它在很多任务中（如文本分类）非常有效。这就像是把文本的“意义”粗略地量化为组成它的“核心词语”的数量。

```python
# 需要先安装 scikit-learn 库
!pip install scikit-learn
```

```python
from sklearn.feature_extraction.text import CountVectorizer

documents = [
    "This text is about language.",
    "Language is fascinating.",
    "Linguistics and language are related."
]
```

```python
# 创建一个 CountVectorizer 对象
# CountVectorizer 会自动进行分词（简单的空格分词）并建立词汇表
vectorizer = CountVectorizer()
```

```python
# 学习词汇表并创建词袋向量
X = vectorizer.fit_transform(documents)

# 查看词汇表
print("词汇表:", vectorizer.get_feature_names_out())
```

```python
# 查看词袋向量 (输出的是稀疏矩阵，toarray() 转换为密集矩阵方便查看)
print("词袋向量:\n", X.toarray())
```

## 命名实体识别 (NER)

**目标：** 学习如何识别文本中的命名实体，如人名、地名、组织名等。

命名实体识别（Named Entity Recognition, NER）是自然语言处理中的一个重要任务，它的目标是识别文本中具有特定意义的实体，通常包括人名、地名、组织名、日期、时间等。

在语言学中，识别命名实体对于信息抽取、关系抽取等任务非常重要。例如，在分析新闻报道时，识别出其中提到的人物、地点和组织，可以帮助我们更好地理解事件。

我们将使用一个更强大的Python库 `spaCy` 来进行NER。spaCy是一个专门用于高级自然语言处理的库，速度快且功能强大。

**Python操作：** 使用 spaCy 库进行命名实体识别。

**语言学意义：** NER就像是给文本中的特定“专有名词”打上标签，帮助我们快速识别出文本中涉及到的具体“实体”。这对于信息抽取和知识图谱构建非常有用。

```python
# 需要先安装 spaCy 库
!pip install spacy

# 下载英文模型
!python -m spacy download en_core_web_sm
```

```python
import spacy

# 加载英文模型
nlp = spacy.load("en_core_web_sm")

text_for_ner = "Barack Obama visited Beijing in 2009. He met with Chinese leaders at the Great Hall of the People."
```

```python
# 使用模型处理文本
doc = nlp(text_for_ner)

# 遍历识别出的命名实体
print("命名实体识别结果:")
for ent in doc.ents:
    print(f"实体: {ent.text}, 类型: {ent.label_}")
```

```python
# 解释实体类型
print("\n常见实体类型解释 (部分):")
print("PERSON: 人名")
print("GPE: 地缘政治实体 (国家、城市等)")
print("ORG: 组织机构")
print("DATE: 日期")
print("LOC: 地点")
```

## 情感分析

**目标：** 初步了解如何使用Python判断文本的情感倾向（积极、消极、中立）。


情感分析（Sentiment Analysis），也称为意见挖掘（Opinion Mining），是判断文本所表达情感（如积极、消极、中立、愤怒、喜悦等）的过程。

在语言学中，情感分析与语用学、语义学等领域密切相关，帮助我们理解文本中隐藏的态度和情感信息。例如，分析用户评论、社交媒体帖子等，可以了解人们对某个产品、事件或观点的看法。

我们将使用 `VADER` (Valence Aware Dictionary and sEntiment Reasoner) 工具进行情感分析。VADER是一个基于词典和规则的情感分析工具，对社交媒体文本特别有效，且无需训练模型。

**Python操作：** 使用 NLTK 库中的 VADER 工具进行情感分析。

**语言学意义：** 情感分析试图捕捉文本中的“情感色彩”或“评价倾向”，这与我们对词语“褒贬义”、语气、语调等语言学概念的理解是相通的。

```python
# 需要下载 vader_lexicon 数据
nltk.download('vader_lexicon')
```

```python
from nltk.sentiment.vader import SentimentIntensityAnalyzer

# 创建一个 SentimentIntensityAnalyzer 对象
analyzer = SentimentIntensityAnalyzer()

sentences = [
    "Python is a great tool for linguistics!", # 积极
    "This task is a bit challenging.", # 中性/轻微消极
    "I am very disappointed with the result.", # 消极
    "This is the best course ever!" # 积极
]
```

```python
print("情感分析结果:")
for sentence in sentences:
    # 获取情感得分
    sentiment_scores = analyzer.polarity_scores(sentence)
    print(f"文本: '{sentence}'")
    print(f"  得分: {sentiment_scores}")
    # 解释得分含义
    # compound 得分范围在 -1 (最消极) 到 +1 (最积极) 之间
    if sentiment_scores['compound'] >= 0.05:
        print("  情感: 积极")
    elif sentiment_scores['compound'] <= -0.05:
        print("  情感: 消极")
    else:
        print("  情感: 中立")
    print("-" * 20)
```
