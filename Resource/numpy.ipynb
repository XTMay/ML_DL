numpy.clip 是一个非常实用的函数，它的作用是 将数组中的元素限制在指定的区间范围内。超出范围的值会被截断到最小值或最大值。

应用场景

	1.	防止数值溢出或下溢（如梯度裁剪、图像处理）
	2.	限制信号范围（如 EEG/pupil 信号归一化）
	3.	图像像素值处理（将像素限制在 [0,255]）


````python
import numpy as np

arr = np.array([1, 5, 70, 15, 20])

# 限制在区间 [5, 15]
clipped = np.clip(arr, 5, 18)
print(clipped)
````

````python
import numpy as np

arr = np.array([50, 10, 30, 20])

# 普通排序
sorted_arr = np.sort(arr)
print(sorted_arr)  
# 输出: [10 20 30 50]

# argsort
indices = np.argsort(arr)
print(indices)
# 输出: [1 3 2 0]

# 用 argsort 索引获取排序结果
print(arr[indices])
# 输出: [10 20 30 50]  → 和 np.sort(arr) 一样
````

np.linalg.solve 是 NumPy 中用来 解线性方程组 的函数。它的作用是求解 Ax = b 中的 x，其中 A 是系数矩阵，b 是常数向量（或矩阵）

````python
import numpy as np

# 系数矩阵 A
A = np.array([[2, 1],
              [3, 4]])

# 常数向量 b
b = np.array([5, 6])

# 求解 x
x = np.linalg.solve(A, b)
print(x)
````

基础用法：条件选择元素

````python
import numpy as np

arr = np.array([1, 2, 3, 4, 5, 6])

# 找出大于3的元素索引
indices = np.where(arr > 3)
print(indices)          # 输出 (array([3, 4, 5]),)
print(arr[indices])     # 输出 [4 5 6]

# 直接用布尔条件
print(arr[arr > 3])     # 同上
````

用 where 做元素替换

````python
arr = np.array([1, 2, 3, 4, 5, 6])

# 小于等于3的元素变成0，大于3的保持原值
new_arr = np.where(arr <= 2, 99, 88)
print(new_arr)  # 输出 [0 0 0 4 5 6]
````

二维数组条件选择

多条件使用 where

````python
arr = np.array([10, 15, 20, 25, 30])

# 条件1：小于20 -> 0
# 条件2：大于等于20 -> 1
new_arr = np.where(arr < 20, 0, 1)
print(new_arr)  # 输出 [0 0 1 1 1]

# 可以嵌套 np.where 实现多条件
# 小于15 -> 0, 15~25 -> 1, 大于25 -> 2
new_arr2 = np.where(arr < 15, 0, np.where(arr <= 25, 1, 2)) # if arr < 15, elif arr <= 25 else
print(new_arr2)  # 输出 [0 1 1 1 2]
````

实际应用：选择两个数组的元素

````python
a = np.array([1, 2, 3, 4, 5])
b = np.array([10, 20, 30, 40, 50])

# 条件：a>3 选择 b，否则选择 a
result = np.where(a > 3, b, a)
print(result)  # 输出 [ 1  2  3 40 50]
````
