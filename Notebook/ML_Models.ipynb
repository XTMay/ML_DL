## Types of Machine Learning


### 1. Supervised Learning
- **Goal**: Learn mapping from input (X) to output (y)
- **Data**: Labeled examples (input-output pairs)
- **Types**:
  - **Classification**: Predict categories (spam/not spam)
  - **Regression**: Predict continuous values (house prices)

### 2. Unsupervised Learning
- **Goal**: Find hidden patterns in data
- **Data**: Unlabeled examples (only inputs)
- **Types**:
  - **Clustering**: Group similar data points
  - **Dimensionality Reduction**: Simplify data while preserving information
  - **Association Rules**: Find relationships between variables

### 3. Reinforcement Learning
- **Goal**: Learn optimal actions through trial and error
- **Data**: Rewards/penalties from environment interactions
- **Examples**: Game playing, robotics, autonomous vehicles

## Why Data Preprocessing?

Real-world data is messy:
- **Missing values**: Incomplete records
- **Different scales**: Age (0-100) vs Income (0-100,000)
- **Categorical data**: Text labels that need numerical encoding
- **Outliers**: Extreme values that can skew models

image.png

## Core Preprocessing Steps

### 1. Handling Missing Values
- **Removal**: Drop rows/columns with missing data
- **Imputation**: Fill with mean, median, mode, or forward fill
- **Indicator**: Create binary flags for missing values

### 2. Feature Scaling
- **Standardization**: Mean=0, Std=1 → $z = \frac{x - \mu}{\sigma}$
- **Normalization**: Scale to [0,1] → $x' = \frac{x - x_{min}}{x_{max} - x_{min}}$
- **Robust Scaling**: Use median and IQR (less sensitive to outliers)

### 3. Categorical Encoding
- **Label Encoding**: Map categories to integers (0, 1, 2, ...)
- **One-Hot Encoding**: Create binary columns for each category
- **Target Encoding**: Replace with target variable statistics

### 4. Train-Test Split
Mathematical principle: $D = D_{train} \cup D_{test}$, where $D_{train} \cap D_{test} = \emptyset$

Common splits:
- 70/30 or 80/20 for train/test
- 60/20/20 for train/validation/test

image.png

## Key Takeaways
- Clean data is crucial for model performance
- Choose preprocessing based on data characteristics
- Always fit preprocessing on training data only
- Use pipelines to avoid data leakage

## Try It Yourself
1. Create a dataset with missing values and handle them
2. Apply different scaling methods and compare results

````python
# Create a messy dataset for preprocessing demonstration
np.random.seed(42)

# Generate sample data
n_samples = 200
data = {
    'age': np.random.randint(18, 70, n_samples),
    'income': np.random.normal(50000, 15000, n_samples),
    'education': np.random.choice(['High School', 'Bachelor', 'Master', 'PhD'], n_samples),
    'experience': np.random.randint(0, 30, n_samples),
    'city': np.random.choice(['NYC', 'LA', 'Chicago', 'Houston'], n_samples)
}

df_raw = pd.DataFrame(data)

# Introduce missing values randomly
missing_indices = np.random.choice(n_samples, 30, replace=False)
df_raw.loc[missing_indices[:15], 'income'] = np.nan
df_raw.loc[missing_indices[15:], 'education'] = np.nan

print("Raw Dataset Overview:")
print(f"Shape: {df_raw.shape}")
print(f"\nMissing values per column:")
print(df_raw.isnull().sum())
print(f"\nFirst 5 rows:")
print(df_raw.head())
````

````python
# Step 1: Handle Missing Values
from sklearn.impute import SimpleImputer

# Create a copy for processing
df_processed = df_raw.copy()

# Impute numerical missing values with median
num_imputer = SimpleImputer(strategy='median')
df_processed['income'] = num_imputer.fit_transform(df_processed[['income']]).ravel()

# Impute categorical missing values with most frequent
cat_imputer = SimpleImputer(strategy='most_frequent')
df_processed['education'] = cat_imputer.fit_transform(df_processed[['education']]).ravel()

print("After handling missing values:")
print(f"Missing values: {df_processed.isnull().sum().sum()}")
print(f"\nIncome statistics before/after imputation:")
print(f"Original mean: {df_raw['income'].mean():.2f}")
print(f"After imputation: {df_processed['income'].mean():.2f}")
````

````python
# Step 2: Categorical Encoding
from sklearn.preprocessing import LabelEncoder, OneHotEncoder

# Label Encoding for ordinal data (education has natural order)
education_order = {'High School': 0, 'Bachelor': 1, 'Master': 2, 'PhD': 3}
df_processed['education_encoded'] = df_processed['education'].map(education_order)

# One-Hot Encoding for nominal data (city has no natural order)
city_encoded = pd.get_dummies(df_processed['city'], prefix='city')
df_encoded = pd.concat([df_processed, city_encoded], axis=1)

print("Categorical Encoding Results:")
print(f"Education mapping: {education_order}")
print(f"\nOriginal education values: {df_processed['education'].unique()}")
print(f"Encoded education values: {df_processed['education_encoded'].unique()}")
print(f"\nOne-hot encoded city columns: {list(city_encoded.columns)}")
print(f"\nFinal dataset shape: {df_encoded.shape}")
````

````python
# Step 3: Feature Scaling Comparison
from sklearn.preprocessing import StandardScaler, MinMaxScaler, RobustScaler

# Select numerical features for scaling
numerical_features = ['age', 'income', 'experience']
X_numerical = df_processed[numerical_features]

# Apply different scaling methods
scalers = {
    'StandardScaler': StandardScaler(),
    'MinMaxScaler': MinMaxScaler(),
    'RobustScaler': RobustScaler()
}

print("Feature Scaling Comparison:")
print(f"Original data statistics:")
print(X_numerical.describe().round(2))

fig, axes = plt.subplots(2, 2, figsize=(15, 10))
axes = axes.ravel()

# Plot original data
axes[0].boxplot(X_numerical.values, labels=numerical_features)
axes[0].set_title('Original Data')
axes[0].set_ylabel('Value')

# Plot scaled data
for idx, (name, scaler) in enumerate(scalers.items(), 1):
    X_scaled = scaler.fit_transform(X_numerical)
    axes[idx].boxplot(X_scaled, labels=numerical_features)
    axes[idx].set_title(f'{name}')
    axes[idx].set_ylabel('Scaled Value')

    print(f"\n{name} statistics:")
    scaled_df = pd.DataFrame(X_scaled, columns=numerical_features)
    print(scaled_df.describe().round(2))

plt.tight_layout()
plt.show()
````

````python
# Step 4: Train-Test Split and Pipeline
from sklearn.model_selection import train_test_split
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer

# Prepare features and target (create synthetic target)
np.random.seed(42)
# Create target based on age and income (people with higher age+income more likely to be 1)
target_score = (df_processed['age'] / df_processed['age'].max() +
               df_processed['income'] / df_processed['income'].max()) / 2
y = (target_score > np.median(target_score)).astype(int)

# Select features for modeling
numerical_cols = ['age', 'income', 'experience']
categorical_cols = ['education', 'city']

X = df_processed[numerical_cols + categorical_cols]

# Create preprocessing pipeline
numerical_transformer = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='median')),
    ('scaler', StandardScaler())
])

categorical_transformer = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='constant', fill_value='missing')),
    ('onehot', OneHotEncoder(drop='first', handle_unknown='ignore'))
])

# Combine transformers
preprocessor = ColumnTransformer(
    transformers=[
        ('num', numerical_transformer, numerical_cols),
        ('cat', categorical_transformer, categorical_cols)
    ])

# Split data
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42, stratify=y
)

# Apply preprocessing
X_train_processed = preprocessor.fit_transform(X_train)
X_test_processed = preprocessor.transform(X_test)

print("Data Splitting and Pipeline Results:")
print(f"Original data shape: {X.shape}")
print(f"Training set: {X_train_processed.shape}")
print(f"Test set: {X_test_processed.shape}")
print(f"Target distribution - Train: {np.bincount(y_train)}, Test: {np.bincount(y_test)}")
print(f"\nPreprocessing pipeline created successfully!")
print(f"Number of features after preprocessing: {X_train_processed.shape[1]}")
````

## What is Exploratory Data Analysis?

**EDA** is the process of analyzing datasets to summarize main characteristics, often using statistical graphics and other data visualization methods.

### Goals of EDA:
- **Understand data structure**: Shape, types, missing values
- **Identify patterns**: Trends, seasonality, correlations
- **Detect anomalies**: Outliers, inconsistencies
- **Generate hypotheses**: What might influence the target variable?
- **Guide preprocessing**: Which transformations are needed?

image.png

## Types of EDA Visualizations

### Univariate Analysis (Single Variable)
- **Histogram**: Distribution of continuous variables
- **Box plot**: Quartiles, outliers, skewness
- **Bar chart**: Frequency of categories

### Bivariate Analysis (Two Variables)
- **Scatter plot**: Relationship between continuous variables
- **Box plot by category**: Compare distributions across groups
- **Heatmap**: Correlation between variables

### Multivariate Analysis (Multiple Variables)
- **Correlation matrix**: All pairwise correlations
- **Pair plot**: Multiple scatter plots
- **Parallel coordinates**: High-dimensional patterns

## Statistical Measures

- **Central Tendency**: Mean, median, mode
- **Dispersion**: Standard deviation, variance, range
- **Skewness**: $S = \frac{E[(X-\mu)^3]}{\sigma^3}$ (symmetry measure)
- **Kurtosis**: $K = \frac{E[(X-\mu)^4]}{\sigma^4}$ (tail heaviness)

- Visualizations help identify data quality issues

## Mathematical Foundation

Linear regression models the relationship between a dependent variable $y$ and independent variables $X$ as:

$$y = \beta_0 + \beta_1 x_1 + \beta_2 x_2 + ... + \beta_n x_n + \epsilon$$

Where:
- $\beta_0$ = intercept (bias term)
- $\beta_i$ = coefficient for feature $x_i$
- $\epsilon$ = error term

### Matrix Form
$$\mathbf{y} = \mathbf{X}\boldsymbol{\beta} + \boldsymbol{\epsilon}$$


### Least Squares Solution
Minimize Sum of Squared Errors: $SSE = \sum_{i=1}^{n}(y_i - \hat{y}_i)^2$

Optimal coefficients: $\boldsymbol{\beta} = (\mathbf{X}^T\mathbf{X})^{-1}\mathbf{X}^T\mathbf{y}$

image.png

## Key Assumptions
1. **Linearity**: Relationship between X and y is linear
2. **Independence**: Observations are independent
3. **Homoscedasticity**: Constant variance of errors
4. **Normality**: Errors are normally distributed
5. **No multicollinearity**: Features are not highly correlated

## Evaluation Metrics

- **Mean Squared Error (MSE)**: $MSE = \frac{1}{n}\sum_{i=1}^{n}(y_i - \hat{y}_i)^2$
- **Root Mean Squared Error (RMSE)**: $RMSE = \sqrt{MSE}$
- **Mean Absolute Error (MAE)**: $MAE = \frac{1}{n}\sum_{i=1}^{n}|y_i - \hat{y}_i|$
- **R-squared (R²)**: $R^2 = 1 - \frac{SS_{res}}{SS_{tot}}$


## Interpretation

- **R²**: Proportion of variance explained by the model (0-1, higher is better)
- **P-values**: Statistical significance of coefficients


## Key Takeaways

# 6: Polynomial Regression

## Learning Objectives
- Understand when and why to use polynomial regression
- Implement polynomial features for non-linear relationships
- Recognize and address overfitting in polynomial models
- Learn about bias-variance tradeoff

## Beyond Linear Relationships

Real-world relationships are often non-linear. Polynomial regression extends linear regression to capture curved relationships:


### Polynomial Features
For a single feature $x$, polynomial regression of degree $d$ creates:

$$y = \beta_0 + \beta_1 x + \beta_2 x^2 + \beta_3 x^3 + ... + \beta_d x^d$$

For multiple features, it includes interaction terms:
$$y = \beta_0 + \beta_1 x_1 + \beta_2 x_2 + \beta_3 x_1^2 + \beta_4 x_2^2 + \beta_5 x_1 x_2 + ...$$

image.png

## Key Concepts

### Bias-Variance Tradeoff
- **Bias**: Error from oversimplifying the model
- **Variance**: Error from sensitivity to small changes in training data
- **Total Error**: $Error = Bias^2 + Variance + Noise$

image.png

### Model Complexity
- **Low degree (1-2)**: High bias, low variance (underfitting)
- **Medium degree (3-5)**: Balanced bias-variance
- **High degree (>5)**: Low bias, high variance (overfitting)

image.png

### Regularization Preview
When polynomial features create overfitting:
- Add penalty term to cost function
- Ridge: $Cost = MSE + \alpha \sum \beta_i^2$
- Lasso: $Cost = MSE + \alpha \sum |\beta_i|$

## Signs of Overfitting
- Training error << Test error
- Model performs well on training data but poorly on new data
- Validation curves show diverging train/test performance
- Very high/low coefficient values

## Key Takeaways
- Polynomial regression captures non-linear patterns
- Higher degrees increase model flexibility but risk overfitting
- Always validate on unseen data
- Feature scaling becomes more important with polynomials

````python
# Generate synthetic non-linear dataset
np.random.seed(42)
n_samples = 100
X_nonlinear = np.linspace(0, 4, n_samples).reshape(-1, 1)

# True function: quadratic with noise
def true_function(x):
    return 0.5 * x**2 + 0.3 * x + 0.2 * np.sin(10 * x)

y_nonlinear = true_function(X_nonlinear.ravel()) + np.random.normal(0, 0.3, n_samples)

# Split data
X_train_nl, X_test_nl, y_train_nl, y_test_nl = train_test_split(
    X_nonlinear, y_nonlinear, test_size=0.3, random_state=42
)

print("Non-linear Dataset Created:")
print(f"Training set: {X_train_nl.shape}")
print(f"Test set: {X_test_nl.shape}")

# Plot the true function and data
plt.figure(figsize=(12, 6))

# Left plot: Data and true function
plt.subplot(1, 2, 1)
x_plot = np.linspace(0, 4, 200)
y_true_plot = true_function(x_plot)
plt.plot(x_plot, y_true_plot, 'r-', linewidth=2, label='True Function')
plt.scatter(X_train_nl, y_train_nl, alpha=0.7, s=50, label='Training Data')
plt.scatter(X_test_nl, y_test_nl, alpha=0.7, s=50, color='orange', label='Test Data')
plt.xlabel('X')
plt.ylabel('y')
plt.title('Non-linear Dataset')
plt.legend()
plt.grid(True, alpha=0.3)

# Right plot: Linear fit (for comparison)
plt.subplot(1, 2, 2)
linear_model = LinearRegression()
linear_model.fit(X_train_nl, y_train_nl)
y_linear_plot = linear_model.predict(x_plot.reshape(-1, 1))

plt.plot(x_plot, y_true_plot, 'r-', linewidth=2, label='True Function')
plt.plot(x_plot, y_linear_plot, 'b--', linewidth=2, label='Linear Fit')
plt.scatter(X_train_nl, y_train_nl, alpha=0.7, s=50, label='Training Data')
plt.xlabel('X')
plt.ylabel('y')
plt.title('Linear Model (Underfitting)')
plt.legend()
plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

# Linear model performance
y_pred_linear = linear_model.predict(X_test_nl)
linear_r2 = r2_score(y_test_nl, y_pred_linear)
linear_rmse = np.sqrt(mean_squared_error(y_test_nl, y_pred_linear))
print(f"\nLinear Model Performance:")
print(f"R²: {linear_r2:.3f}, RMSE: {linear_rmse:.3f}")
````

````python
# Compare different polynomial degrees
from sklearn.preprocessing import PolynomialFeatures
from sklearn.pipeline import Pipeline
from sklearn.metrics import mean_squared_error, r2_score

degrees = [1, 2, 3, 5, 10, 15]
results = []

plt.figure(figsize=(18, 12))

for idx, degree in enumerate(degrees):
    # Create polynomial pipeline
    poly_pipeline = Pipeline([
        ('poly', PolynomialFeatures(degree=degree, include_bias=False)),
        ('scaler', StandardScaler()),
        ('linear', LinearRegression())
    ])

    # Fit model
    poly_pipeline.fit(X_train_nl, y_train_nl)

    # Predictions
    y_train_pred = poly_pipeline.predict(X_train_nl)
    y_test_pred = poly_pipeline.predict(X_test_nl)

    # Calculate metrics
    train_r2 = r2_score(y_train_nl, y_train_pred)
    test_r2 = r2_score(y_test_nl, y_test_pred)
    train_rmse = np.sqrt(mean_squared_error(y_train_nl, y_train_pred))
    test_rmse = np.sqrt(mean_squared_error(y_test_nl, y_test_pred))

    results.append({
        'degree': degree,
        'train_r2': train_r2,
        'test_r2': test_r2,
        'train_rmse': train_rmse,
        'test_rmse': test_rmse,
        'gap': train_r2 - test_r2
    })

    # Plot predictions
    plt.subplot(2, 3, idx + 1)
    x_plot = np.linspace(0, 4, 200)
    y_pred_plot = poly_pipeline.predict(x_plot.reshape(-1, 1))

    plt.plot(x_plot, true_function(x_plot), 'r-', linewidth=2, label='True Function')
    plt.plot(x_plot, y_pred_plot, 'b--', linewidth=2, label=f'Polynomial Degree {degree}')
    plt.scatter(X_train_nl, y_train_nl, alpha=0.7, s=30, color='blue', label='Training')
    plt.scatter(X_test_nl, y_test_nl, alpha=0.7, s=30, color='orange', label='Test')

    plt.title(f'Degree {degree}\nTrain R²: {train_r2:.3f}, Test R²: {test_r2:.3f}')
    plt.xlabel('X')
    plt.ylabel('y')
    plt.legend(fontsize=8)
    plt.grid(True, alpha=0.3)

    # Limit y-axis for very high degrees to show reasonable range
    if degree >= 10:
        plt.ylim(-2, 4)

plt.tight_layout()
plt.show()

# Results summary
results_df = pd.DataFrame(results)
print("\nPolynomial Regression Results Summary:")
print(results_df.round(3))
````

````python
# Learning Curves - Bias-Variance Tradeoff Visualization
plt.figure(figsize=(15, 5))

# 1. Performance vs Degree
plt.subplot(1, 3, 1)
plt.plot(results_df['degree'], results_df['train_r2'], 'bo-', label='Training R²', linewidth=2)
plt.plot(results_df['degree'], results_df['test_r2'], 'ro-', label='Test R²', linewidth=2)
plt.xlabel('Polynomial Degree')
plt.ylabel('R² Score')
plt.title('Model Performance vs Complexity')
plt.legend()
plt.grid(True, alpha=0.3)

# Mark sweet spot
best_degree = results_df.loc[results_df['test_r2'].idxmax(), 'degree']
best_test_r2 = results_df['test_r2'].max()
plt.axvline(x=best_degree, color='green', linestyle='--', alpha=0.7)
plt.text(best_degree + 0.5, best_test_r2 - 0.1, f'Best: {best_degree}',
         bbox=dict(boxstyle='round', facecolor='lightgreen'))

# 2. RMSE vs Degree
plt.subplot(1, 3, 2)
plt.plot(results_df['degree'], results_df['train_rmse'], 'bo-', label='Training RMSE', linewidth=2)
plt.plot(results_df['degree'], results_df['test_rmse'], 'ro-', label='Test RMSE', linewidth=2)
plt.xlabel('Polynomial Degree')
plt.ylabel('RMSE')
plt.title('Error vs Complexity')
plt.legend()
plt.grid(True, alpha=0.3)
plt.axvline(x=best_degree, color='green', linestyle='--', alpha=0.7)

# 3. Overfitting Gap
plt.subplot(1, 3, 3)
plt.plot(results_df['degree'], results_df['gap'], 'mo-', linewidth=2)
plt.axhline(y=0, color='black', linestyle='-', alpha=0.3)
plt.axhline(y=0.1, color='red', linestyle='--', alpha=0.7, label='Overfitting Threshold')
plt.xlabel('Polynomial Degree')
plt.ylabel('Training R² - Test R²')
plt.title('Overfitting Gap')
plt.legend()
plt.grid(True, alpha=0.3)
plt.axvline(x=best_degree, color='green', linestyle='--', alpha=0.7)

plt.tight_layout()
plt.show()

# Analysis
print("\nBias-Variance Analysis:")
print(f"Best polynomial degree: {best_degree} (Test R²: {best_test_r2:.3f})")

for _, row in results_df.iterrows():
    if row['degree'] == 1:
        status = "Underfitting (High Bias)"
    elif row['gap'] > 0.1:
        status = "Overfitting (High Variance)"
    elif row['degree'] == best_degree:
        status = "Good Balance (Sweet Spot)"
    else:
        status = "Reasonable"

    print(f"Degree {int(row['degree']):2d}: {status:30} | Gap: {row['gap']:+.3f}")

print(f"\nKey Insights:")
print(f"• Degree 1 (linear): High bias, captures linear trend only")
print(f"• Degree {best_degree}: Best generalization, captures non-linearity without overfitting")
print(f"• High degrees (>5): Low bias but high variance, memorize noise")
print(f"• Sweet spot balances complexity and generalization")
````

# 7: Logistic Regression

## Learning Objectives
- Understand logistic regression for binary and multiclass classification
- Learn the sigmoid function and maximum likelihood estimation
- Implement and evaluate logistic regression models
- Interpret coefficients and probabilities in classification context

## From Linear to Logistic Regression

Linear regression predicts continuous values, but classification requires predicting probabilities between 0 and 1.

image.png

### The Sigmoid Function
Transform linear output to probability using sigmoid:

$$\sigma(z) = \frac{1}{1 + e^{-z}} = \frac{e^z}{1 + e^z}$$

Where: $z = \beta_0 + \beta_1 x_1 + \beta_2 x_2 + ... + \beta_n x_n$

### Logistic Regression Model
$$P(y=1|X) = \frac{1}{1 + e^{-(\beta_0 + \beta_1 x_1 + ... + \beta_n x_n)}}$$


### Odds and Log-Odds
- **Odds**: $\text{Odds} = \frac{P(y=1)}{P(y=0)} = \frac{P(y=1)}{1-P(y=1)}$
- **Log-Odds (Logit)**: $\log(\text{Odds}) = \beta_0 + \beta_1 x_1 + ... + \beta_n x_n$

image.png

## Maximum Likelihood Estimation

Unlike linear regression (least squares), logistic regression uses MLE:

**Likelihood Function**: $L(\beta) = \prod_{i=1}^{n} P(y_i|x_i, \beta)$

**Log-Likelihood**: $\ell(\beta) = \sum_{i=1}^{n} [y_i \log(p_i) + (1-y_i) \log(1-p_i)]$

Goal: Maximize log-likelihood (minimize negative log-likelihood)

Maximum Likelihood Estimation (MLE)

1. Introduction

	•	Problem: How to estimate unknown parameters from data?
  
	•	MLE: Choose parameters that maximize the probability of observing the data.


2. Theory

Likelihood Function

For independent data
$$y_1, y_2, \dots, y_n
:$$

$$
L(\theta) = \prod_{i=1}^n P(y_i \mid \theta)
$$

MLE is:

$$
\hat{\theta}{\mathrm{MLE}} = \arg\max{\theta} L(\theta)
$$

Log-likelihood:

$$
\ell(\theta) = \sum_{i=1}^n \log P(y_i \mid \theta)
$$

Example: Coin Toss

Observed: k heads out of n tosses.

$$
L(p) = p^k (1-p)^{,n-k}
$$

$$
\ell(p) = k \log p + (n-k),\log(1-p)
$$

$$
\hat{p} = \frac{k}{n}
$$



````python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

# ============================
# 1. Coin Toss MLE
# ============================

# Simulate coin tosses: true probability of heads
p_true = 0.7
n = 20
data = np.random.binomial(1, p_true, size=n)  # 1=head, 0=tail
print("Coin toss results:", data)

# Likelihood function: L(p) = p^k * (1-p)^(n-k)
k = data.sum()  # number of heads
def likelihood(p):
    return p**k * (1-p)**(n-k)

# Scan possible p values to find the maximum likelihood estimate
p_vals = np.linspace(0.01, 0.99, 100)
likelihoods = [likelihood(p) for p in p_vals]
p_mle = p_vals[np.argmax(likelihoods)]

print(f"MLE estimate p = {p_mle:.3f}, True p = {p_true}")

plt.plot(p_vals, likelihoods, label="Likelihood")
plt.axvline(p_true, color='r', linestyle="--", label="True p")
plt.axvline(p_mle, color='g', linestyle="--", label="MLE estimate")
plt.xlabel("p")
plt.ylabel("Likelihood")
plt.legend()
plt.title("Coin Toss MLE")
plt.show()


# ============================
# 2. Normal Distribution MLE
# ============================

# Simulate normal distribution data
mu_true, sigma_true = 5, 2
data = np.random.normal(mu_true, sigma_true, size=200)

# MLE for normal distribution:
# mu_hat = sample mean
# sigma_hat = sample std (divide by n, not n-1)
mu_mle = np.mean(data)
sigma_mle = np.std(data, ddof=0)  # MLE uses n instead of n-1

print(f"True mu = {mu_true}, sigma = {sigma_true}")
print(f"MLE estimate mu = {mu_mle:.3f}, sigma = {sigma_mle:.3f}")

# Plot histogram and fitted curves
x = np.linspace(mu_true - 4*sigma_true, mu_true + 4*sigma_true, 200)
plt.hist(data, bins=20, density=True, alpha=0.5, label="Data Histogram")
plt.plot(x, norm.pdf(x, mu_true, sigma_true), 'r--', label="True Distribution")
plt.plot(x, norm.pdf(x, mu_mle, sigma_mle), 'g-', label="MLE Fit")
plt.legend()
plt.title("Normal Distribution MLE")
plt.show()
````

Example: Normal Distribution

For data from $$ {N}(\mu, \sigma^2):$$

$$
\hat{\mu} = \frac{1}{n}\sum_{i=1}^n y_i
$$

$$
\hat{\sigma}^{2} = \frac{1}{n}\sum_{i=1}^n \bigl(y_i - \hat{\mu}\bigr)^2
$$


# Derivation: MLE and KL-Divergence

1. **Starting with MLE:**

$$
\hat{\theta} = \arg \max_{\theta} L_{P_{\theta}}(\mathbf{y})
= \arg \max_{\theta} P_{\theta}(\mathbf{y})
= \arg \max_{\theta} P(\mathbf{y}|\theta)
$$

---

2. **Using the product rule for independent observations and converting to log-likelihood:**

$$
= \arg \max_{\theta} \prod_{i=1}^{n} P(y_{i}|\theta)
= \arg \max_{\theta} \sum_{i=1}^{n} \log P(y_{i}|\theta)
$$

---

3. **Introducing a reference distribution $P(y_{i}|\theta_{0})$ and manipulating the expression:**

$$
= \arg \max_{\theta} \left(\sum_{i=1}^{n} \log P(y_{i}|\theta)
- \sum_{i=1}^{n} \log P(y_{i}|\theta_{0})\right)
$$

$$
= \arg \max_{\theta} \sum_{i=1}^{n} \left(\log P(y_{i}|\theta) - \log P(y_{i}|\theta_{0})\right)
$$

---

4. **Rewriting using logarithm properties and converting to argmin:**

$$
= \arg \max_{\theta} \sum_{i=1}^{n} \log \frac{P(y_{i}|\theta)}{P(y_{i}|\theta_{0})}
$$

$$
= \arg \min_{\theta} \sum_{i=1}^{n} \log \frac{P(y_{i}|\theta_{0})}{P(y_{i}|\theta)}
= \arg \min_{\theta} \frac{1}{n} \sum_{i=1}^{n} \log \frac{P(y_{i}|\theta_{0})}{P(y_{i}|\theta)}
$$

---

5. **Defining $h_{\theta}(y_{i})$ and taking the limit as $n \to \infty$:**

$$
= \arg \min_{\theta} \frac{1}{n} \sum_{i=1}^{n} h_{\theta}(y_{i})
\xrightarrow{n \to \infty}
\arg \min_{\theta} \, E[h_{\theta}(y)]
$$

---

6. **Expressing the expectation as an integral:**

$$
= \arg \min_{\theta} \int P_{\theta_{0}}(y) \, h_{\theta}(y) \, dy
= \arg \min_{\theta} \int P_{\theta_{0}}(y) \, \log \frac{P(y|\theta_{0})}{P(y|\theta)} \, dy
$$

---

7. **Recognizing the final form as KL-divergence:**

$$
= \arg \min_{\theta} D_{KL}(P_{\theta_{0}} \,||\, P_{\theta})
$$

---

### Conclusion
This derivation shows that **maximizing the likelihood** is equivalent to **minimizing the KL-divergence** between the true data distribution ($P_{\theta_{0}}$) and the model's estimated distribution ($P_{\theta}$).

KL-Divergence (Kullback–Leibler Divergence)

1. Definition

For discrete distributions P and Q:

$$
D_{KL}(P \parallel Q) = \sum_{x} P(x) \log \frac{P(x)}{Q(x)}
$$

For continuous distributions:

$$
D_{KL}(P \parallel Q) = \int P(x) \log \frac{P(x)}{Q(x)} , dx
$$



2. Intuitive Explanation
	•	KL-divergence measures the difference between the true distribution P and the model distribution Q.
	•	Smaller D_{KL} → Q is closer to P.
	•	Not symmetric:

$$
D_{KL}(P \parallel Q) \neq D_{KL}(Q \parallel P)
$$
	•	Interpretation:
If you encode data using Q but the true data follows P, KL-divergence tells you how many extra “bits” on average are needed per data point.



3. Connection to MLE

In Maximum Likelihood Estimation (MLE):
	•	Maximizing likelihood is equivalent to minimizing KL-divergence:

$$
\hat{\theta}{\mathrm{MLE}} = \arg\min{\theta} D_{KL}(P_{\text{true}} \parallel P_\theta)
$$
	•	Intuition: MLE finds parameters so that the model distribution P_\theta is as close as possible to the true data distribution P_{\text{true}}.

## Algorithm Concept

k-NN is a **lazy learning** algorithm that makes predictions based on the k closest training examples:

### Classification
Predict the **most frequent class** among k nearest neighbors

### Regression
Predict the **average value** of k nearest neighbors


image.png

## Distance Metrics

### Euclidean Distance (L2)

$$
d(x_i, x_j) = \ sqrt {\sum_{k=1}^{n} (x_{ik} - x_{jk})^2}
$$


### Manhattan Distance (L1)
$$d(x_i, x_j) = \sum_{k=1}^{n}|x_{ik} - x_{jk}|$$

### Minkowski Distance (General)
$$d(x_i, x_j) = \left(\sum_{k=1}^{n}|x_{ik} - x_{jk}|^p\right)^{1/p}$$

Where $p=1$ (Manhattan), $p=2$ (Euclidean)


image.png

## Key Properties

### Advantages
- **Simple**: No assumptions about data distribution
- **Versatile**: Works for classification and regression
- **Non-parametric**: Adapts to local patterns
- **Interpretable**: Easy to understand predictions

### Disadvantages
- **Computationally expensive**: Must compute all distances
- **Memory intensive**: Stores entire training set
- **Sensitive to irrelevant features**: Curse of dimensionality
- **Sensitive to scale**: Requires feature scaling


## Choosing k

- **Small k (k=1)**: Low bias, high variance (overfitting)
- **Large k**: High bias, low variance (underfitting)
- **Rule of thumb**: $k = \sqrt{n}$ where n is number of samples
- **Cross-validation**: Find optimal k empirically
- **Odd k**: Avoids ties in binary classification


http://vision.stanford.edu/teaching/cs231n-demos/knn/

## Curse of Dimensionality

In high dimensions:
- All points become equidistant
- "Nearest" neighbors are no longer meaningful
- Performance degrades significantly


## Key Takeaways
- k-NN is simple but can be very effective
- Feature scaling is crucial
- Works well with small to medium datasets
- Poor performance in high dimensions
- No training phase, but expensive prediction

## Bayes' Theorem Foundation

Bayes' theorem describes probability of an event based on prior knowledge:

$$P(A|B) = \frac{P(B|A) \cdot P(A)}{P(B)}$$

### For Classification
$$P(C_k|X) = \frac{P(X|C_k) \cdot P(C_k)}{P(X)}$$

Where:
- $P(C_k|X)$ = Posterior probability (what we want)
- $P(X|C_k)$ = Likelihood (probability of features given class)
- $P(C_k)$ = Prior probability (class frequency)
- $P(X)$ = Evidence (marginal probability, constant for all classes)

## The "Naïve" Assumption

**Independence Assumption**: Features are conditionally independent given the class

$$P(X|C_k) = P(x_1, x_2, ..., x_n|C_k) = \prod_{i=1}^{n} P(x_i|C_k)$$

This transforms complex multivariate problem into simpler univariate problems.


## Naïve Bayes Variants

### 1. Gaussian Naïve Bayes
For continuous features following normal distribution:
$$P(x_i|C_k) = \frac{1}{\sqrt{2\pi\sigma_{k}^2}} e^{-\frac{(x_i-\mu_k)^2}{2\sigma_k^2}}$$

### 2. Multinomial Naïve Bayes
For discrete count data (text, word frequencies):
$$P(x_i|C_k) = \frac{N_{ki} + \alpha}{N_k + \alpha n}$$
Where $N_{ki}$ = count of feature $i$ in class $k$, $\alpha$ = smoothing parameter

### 3. Bernoulli Naïve Bayes
For binary features (presence/absence):
$$P(x_i|C_k) = P_{ki}x_i + (1-P_{ki})(1-x_i)$$


## Key Advantages & Applications
- **Fast**: Linear time complexity
- **Simple**: Few parameters to estimate
- **Robust**: Works well with small training sets
- **Text Classification**: Excellent for spam filtering, sentiment analysis
- **Medical Diagnosis**: Symptom-based classification

## Key Takeaways
- Despite "naïve" assumption, often performs surprisingly well
- Excellent baseline model for comparison
- Fast training and prediction
- Good for high-dimensional data

# 10: Support Vector Machines (SVM)


## Learning Objectives
- Understand the concept of maximum margin classification
- Learn about support vectors and the kernel trick
- Implement SVMs for linear and non-linear classification
- Recognize when SVMs are most effective

## The Maximum Margin Concept

SVM finds the optimal hyperplane that separates classes with the **maximum margin**.

### Key Definitions
- **Hyperplane**: Decision boundary that separates classes
- **Support Vectors**: Data points closest to the hyperplane
- **Margin**: Distance between hyperplane and nearest data points

### Mathematical Formulation
For a linear SVM, the hyperplane is defined by:
$$w^T x + b = 0$$

The optimization problem is:
$$\min_{w,b} \frac{1}{2}||w||^2$$

Subject to: $y_i(w^T x_i + b) \geq 1$ for all $i$

## The Kernel Trick

For non-linearly separable data, SVM uses kernels to map data to higher dimensions:

$$K(x_i, x_j) = \phi(x_i)^T \phi(x_j)$$

### Common Kernels

#### Linear Kernel
$$K(x_i, x_j) = x_i^T x_j$$

#### Polynomial Kernel
$$K(x_i, x_j) = (x_i^T x_j + c)^d$$

#### Radial Basis Function (RBF/Gaussian) Kernel
$$K(x_i, x_j) = \exp(-\gamma ||x_i - x_j||^2)$$

#### Sigmoid Kernel
$$K(x_i, x_j) = \tanh(\gamma x_i^T x_j + c)$$

## Soft Margin SVM

For non-separable data, introduce slack variables $\xi_i$:

$$\min_{w,b,\xi} \frac{1}{2}||w||^2 + C\sum_{i=1}^{n}\xi_i$$

Subject to: $y_i(w^T x_i + b) \geq 1 - \xi_i$ and $\xi_i \geq 0$

- **C parameter**: Controls trade-off between margin maximization and classification errors
- **High C**: Hard margin (less tolerance for misclassification)
- **Low C**: Soft margin (more tolerance for misclassification)


## Key Properties

### Advantages
- **Effective in high dimensions**: Works well with many features
- **Memory efficient**: Only uses support vectors
- **Versatile**: Different kernels for different problems
- **Works with small datasets**: No assumptions about data distribution

### Disadvantages
- **Slow on large datasets**: O(n³) training complexity
- **Sensitive to feature scaling**: Requires normalization
- **No probability estimates**: Only gives class predictions
- **Difficult to interpret**: Black box model

## Applications
- **Text classification**: Document categorization
- **Image classification**: Object recognition
- **Bioinformatics**: Gene classification
- **Financial**: Credit scoring

## Key Takeaways
- SVM finds optimal separating hyperplane with maximum margin
- Kernel trick enables non-linear classification
- C parameter controls overfitting vs underfitting
- Excellent for high-dimensional problems

## Learning Objectives

- Learn about entropy, information gain, and Gini impurity
- Implement decision trees for classification and regression

## Decision Tree Concept

A decision tree is a flowchart-like structure that makes decisions by asking a series of questions about the features.

### Key Components
- **Root Node**: Starting point (entire dataset)
- **Internal Nodes**: Decision points (questions about features)
- **Leaf Nodes**: Final predictions (class or value)
- **Branches**: Outcomes of decisions


### Decision Process
1. Start at root with all data
2. Find best feature and threshold to split on
3. Create branches for each outcome
4. Repeat recursively until stopping criteria met

## Splitting Criteria

### For Classification

#### Entropy
Measures randomness/uncertainty in the data:
$$H(S) = -\sum_{i=1}^{c} p_i \log_2(p_i)$$

Where $p_i$ is the proportion of samples belonging to class $i$.

#### Information Gain
Reduction in entropy after splitting:
$$IG(S, A) = H(S) - \sum_{v \in Values(A)} \frac{|S_v|}{|S|} H(S_v)$$

#### Gini Impurity
Alternative to entropy:
$$Gini(S) = 1 - \sum_{i=1}^{c} p_i^2$$


### For Regression

#### Mean Squared Error (MSE)
$$MSE = \frac{1}{n}\sum_{i=1}^{n}(y_i - \bar{y})^2$$

#### Mean Absolute Error (MAE)
$$MAE = \frac{1}{n}\sum_{i=1}^{n}|y_i - \bar{y}|$$

## Tree Building Algorithm (CART)

1. **Select best split**: Choose feature and threshold that maximizes information gain
2. **Create branches**: Split data based on chosen criteria
3. **Recurse**: Apply same process to each subset
4. **Stop when**:
   - Maximum depth reached
   - Minimum samples per leaf reached
   - No improvement in purity

## Advantages & Disadvantages

### Advantages
- **Interpretable**: Easy to understand and visualize
- **No assumptions**: Works with any data distribution
- **Handles mixed data**: Both numerical and categorical features
- **Feature selection**: Automatically identifies important features
- **Non-linear**: Captures complex patterns

### Disadvantages
- **Overfitting**: Tends to memorize training data
- **Instability**: Small changes in data can drastically change tree
- **Bias**: Favors features with more levels
- **Difficulty with linear relationships**: May create complex splits

## Overfitting and Pruning

### Pre-pruning (Early Stopping)
- **Max depth**: Limit tree depth
- **Min samples split**: Minimum samples required to split
- **Min samples leaf**: Minimum samples in leaf nodes
- **Min impurity decrease**: Minimum improvement required

### Post-pruning
- Build full tree then remove branches
- **Cost complexity pruning**: Balance accuracy vs tree size
- **Cross-validation**: Use validation set to determine optimal size

## Key Takeaways
- Decision trees create interpretable models through recursive splitting
- Information gain and Gini impurity guide splitting decisions
- Prone to overfitting without proper regularization
- Foundation for ensemble methods (Random Forests, Gradient Boosting)

# 12: Random Forests

## Learning Objectives
- Understand ensemble learning and bootstrap aggregating (bagging)
- Learn how Random Forests improve upon decision trees
- Implement Random Forests for classification and regression
- Analyze feature importance and out-of-bag error estimation

## Ensemble Learning Concept

**Ensemble Learning**: Combine multiple models to create a stronger predictor than any individual model.

**Key Principle**: "Wisdom of crowds" - diverse models make different errors, which cancel out when averaged.


### Types of Ensemble Methods
- **Bagging**: Train models on different subsets (Bootstrap Aggregating)
- **Boosting**: Train models sequentially, focusing on previous errors
- **Stacking**: Use meta-model to combine predictions

## Random Forest Algorithm

Random Forest combines **bagging** with **random feature selection**.

### Bootstrap Sampling
Create diverse training sets by sampling with replacement:
1. From dataset of size N, sample N examples with replacement
2. Each bootstrap sample has ~63.2% unique examples
3. Train one decision tree on each bootstrap sample

### Random Feature Selection
At each split in each tree:
1. Randomly select $m$ features from total $p$ features
2. Find best split among these $m$ features only
3. Common choices: $m = \sqrt{p}$ (classification), $m = p/3$ (regression)

### Prediction Process

#### Classification
$$\hat{y} = \text{mode}(T_1(x), T_2(x), ..., T_B(x))$$
(Majority vote from all trees)

#### Regression
$$\hat{y} = \frac{1}{B}\sum_{i=1}^{B} T_i(x)$$
(Average of all tree predictions)


## Key Advantages of Random Forests

### Reduces Overfitting
- Individual trees may overfit, but averaging reduces variance
- Random feature selection decorrelates trees

### Out-of-Bag (OOB) Error Estimation
- Each tree trained on ~63% of data
- Remaining ~37% used for validation (OOB samples)
- No need for separate validation set


### Feature Importance
Two methods to calculate importance:


#### Mean Decrease in Impurity
Average reduction in node impurity when splitting on feature across all trees

#### Mean Decrease in Accuracy (Permutation Importance)
Decrease in OOB accuracy when feature values are randomly permuted

### Handles Missing Values
- Can maintain accuracy even with missing data
- Proximity measures help impute missing values


## Hyperparameters

### Key Parameters
- **n_estimators**: Number of trees (more is usually better)
- **max_features**: Number of features considered per split
- **max_depth**: Maximum depth of individual trees
- **min_samples_split**: Minimum samples to split internal node
- **min_samples_leaf**: Minimum samples in leaf node
- **bootstrap**: Whether to use bootstrap samples

### Typical Settings
- **Classification**: max_features = √p
- **Regression**: max_features = p/3
- **n_estimators**: 100-1000 (diminishing returns after certain point)


## Advantages vs Disadvantages

### Advantages
- **Robust**: Less prone to overfitting than individual trees
- **Handles large datasets**: Efficient parallel training
- **Feature importance**: Built-in feature ranking
- **Mixed data types**: Works with numerical and categorical features
- **No scaling needed**: Tree-based, so feature scaling not required
- **OOB validation**: Built-in performance estimation

### Disadvantages
- **Less interpretable**: Harder to understand than single tree
- **Memory intensive**: Stores multiple trees
- **Can overfit**: With very noisy data or too many trees
- **Biased toward categorical features**: With many categories


## Applications
- **Bioinformatics**: Gene expression analysis
- **Finance**: Credit scoring, algorithmic trading
- **E-commerce**: Recommendation systems
- **Computer Vision**: Object detection features
- **Medicine**: Disease prediction

## Key Takeaways
- Random Forests combine bagging with random feature selection
- Excellent balance of accuracy and interpretability
- Built-in cross-validation through OOB error
- Great general-purpose algorithm for many problems
- Foundation for more advanced ensemble methods

````python
# Helper function for plotting confusion matrix
def plot_confusion_matrix(y_true, y_pred, title="Confusion Matrix"):
    cm = confusion_matrix(y_true, y_pred)
    plt.figure(figsize=(8, 6))
    plt.imshow(cm, interpolation='nearest', cmap='Blues')
    plt.title(title)
    plt.colorbar()

    # Add text annotations
    thresh = cm.max() / 2.
    for i in range(cm.shape[0]):
        for j in range(cm.shape[1]):
            plt.text(j, i, format(cm[i, j], 'd'),
                    ha="center", va="center",
                    color="white" if cm[i, j] > thresh else "black")

    plt.ylabel('True Label')
    plt.xlabel('Predicted Label')
    plt.tight_layout()
    plt.show()

# Helper function for feature importance plotting
def plot_feature_importance(model, feature_names, title="Feature Importance"):
    importance = model.feature_importances_
    indices = np.argsort(importance)[::-1]

    plt.figure(figsize=(10, 6))
    plt.title(title)
    plt.bar(range(len(importance)), importance[indices])
    plt.xticks(range(len(importance)), [feature_names[i] for i in indices], rotation=45)
    plt.tight_layout()
    plt.show()

    return importance[indices], [feature_names[i] for i in indices]
````

# 13: Gradient Boosting (XGBoost/LightGBM)


## Learning Objectives
- Understand gradient boosting as sequential ensemble learning
- Learn the difference between bagging and boosting
- Implement XGBoost and LightGBM for classification and regression
- Master hyperparameter tuning for gradient boosting models

## Gradient Boosting Concept

**Boosting**: Sequential ensemble method where each model corrects errors of previous models.

### Key Principle
Build models iteratively, each one focused on the mistakes of the previous ensemble:

$$F_m(x) = F_{m-1}(x) + \gamma_m h_m(x)$$

Where:
- $F_m(x)$ = ensemble prediction at iteration $m$
- $h_m(x)$ = weak learner (decision tree) at iteration $m$
- $\gamma_m$ = learning rate (shrinkage parameter)

## Algorithm Steps

1. **Initialize**: $F_0(x) = \text{constant}$ (mean for regression, log-odds for classification)
2. **For m = 1 to M**:
   - Calculate residuals: $r_{im} = -\frac{\partial L(y_i, F_{m-1}(x_i))}{\partial F_{m-1}(x_i)}$
   - Fit weak learner $h_m(x)$ to residuals
   - Find optimal step size: $\gamma_m = \arg\min_{\gamma} \sum_{i=1}^{n} L(y_i, F_{m-1}(x_i) + \gamma h_m(x_i))$
   - Update: $F_m(x) = F_{m-1}(x) + \gamma_m h_m(x)$

## Loss Functions

### Regression
- **Squared Loss**: $L(y, F(x)) = \frac{1}{2}(y - F(x))^2$
- **Absolute Loss**: $L(y, F(x)) = |y - F(x)|$
- **Huber Loss**: Combination of squared and absolute loss

### Classification
- **Logistic Loss**: $L(y, F(x)) = \log(1 + \exp(-yF(x)))$
- **Exponential Loss**: $L(y, F(x)) = \exp(-yF(x))$ (AdaBoost)

## XGBoost (Extreme Gradient Boosting)

### Key Innovations
- **Regularization**: L1 and L2 penalties to prevent overfitting
- **Tree Pruning**: Bottom-up pruning using gain threshold
- **Parallel Processing**: Efficient parallel tree construction
- **Handling Missing Values**: Built-in missing value handling

### Objective Function
$$Obj^{(t)} = \sum_{i=1}^{n} l(y_i, \hat{y}_i^{(t-1)} + f_t(x_i)) + \Omega(f_t) + \text{constant}$$

Where $\Omega(f_t) = \gamma T + \frac{1}{2}\lambda \sum_{j=1}^{T} w_j^2$ is the regularization term.

## LightGBM (Light Gradient Boosting Machine)

### Key Features
- **Leaf-wise Tree Growth**: Instead of level-wise (faster, more efficient)
- **Gradient-based One-Side Sampling (GOSS)**: Focuses on high-gradient instances
- **Exclusive Feature Bundling (EFB)**: Reduces feature dimensionality
- **Categorical Feature Support**: Direct handling without encoding

## Hyperparameters

### Core Parameters
- **n_estimators**: Number of boosting iterations
- **learning_rate**: Shrinkage factor (0.01-0.3)
- **max_depth**: Maximum tree depth
- **min_child_samples**: Minimum samples per leaf
- **subsample**: Fraction of samples for each tree
- **colsample_bytree**: Fraction of features for each tree

### Regularization Parameters
- **reg_alpha**: L1 regularization
- **reg_lambda**: L2 regularization
- **gamma**: Minimum loss reduction for split

## Advantages vs Disadvantages

### Advantages
- **High Performance**: Often wins ML competitions
- **Feature Importance**: Built-in feature ranking
- **Handles Missing Values**: No preprocessing needed
- **Flexible**: Works for classification and regression
- **Scalable**: Efficient for large datasets

### Disadvantages
- **Overfitting**: Can overfit with wrong hyperparameters
- **Hyperparameter Sensitive**: Requires careful tuning
- **Sequential**: Cannot parallelize across trees (only within)
- **Memory Intensive**: Can consume significant memory

## Boosting vs Bagging

| Aspect | Boosting | Bagging (Random Forest) |
|--------|----------|-------------------------|
| Training | Sequential | Parallel |
| Error Focus | High-error samples | All samples equally |
| Variance | Lower | Higher |
| Bias | Lower | Higher |
| Overfitting Risk | Higher | Lower |
| Speed | Slower | Faster |

## Applications
- **Structured Data**: Tabular data competitions
- **Click-Through Rate**: Online advertising
- **Risk Modeling**: Finance and insurance
- **Ranking**: Search engines and recommendations
- **Feature Selection**: Built-in importance ranking

## Key Takeaways
- Gradient boosting builds strong predictors from weak learners
- XGBoost and LightGBM are optimized implementations
- Sequential nature allows correcting previous errors
- Requires careful hyperparameter tuning
- Excellent performance on structured/tabular data

## Why Dimensionality Reduction?

### The Curse of Dimensionality
As the number of features increases, several problems arise:
- **Exponential Growth**: Data becomes increasingly sparse in high-dimensional space
- **Distance Metrics**: All points become equidistant from each other
- **Computational Cost**: Processing time and memory requirements grow exponentially
- **Visualization**: Impossible to visualize data beyond 3 dimensions

### Benefits of Dimensionality Reduction:
- **Noise Reduction**: Remove irrelevant features and noise
- **Visualization**: Enable 2D/3D plotting of high-dimensional data
- **Storage**: Reduce memory and storage requirements
- **Speed**: Faster training and inference
- **Preprocessing**: Improve performance of downstream algorithms

### Types of Dimensionality Reduction:
1. **Linear**: PCA, LDA, ICA
2. **Non-linear**: t-SNE, UMAP, Autoencoders, Kernel PCA

## Principal Component Analysis (PCA)

PCA finds the directions (principal components) along which the data varies the most.

### Mathematical Foundation:
1. **Standardize** the data: $X_{std} = \frac{X - \mu}{\sigma}$
2. **Compute covariance matrix**: $C = \frac{1}{n-1}X^T X$
3. **Find eigenvectors and eigenvalues**: $C v_i = \lambda_i v_i$
4. **Sort by eigenvalues**: $\lambda_1 \geq \lambda_2 \geq ... \geq \lambda_p$
5. **Project data**: $Y = X W_k$ where $W_k$ contains first $k$ eigenvectors

### Key Properties:
- **Variance Explained**: $\frac{\lambda_i}{\sum_j \lambda_j}$ for component $i$
- **Reconstruction Error**: Minimized by PCA
- **Orthogonal Components**: Principal components are uncorrelated
- **Linear Transformation**: Preserves linear relationships

## t-SNE (t-Distributed Stochastic Neighbor Embedding)

t-SNE is a non-linear dimensionality reduction technique particularly well-suited for visualization.

### Key Concepts:
- **Probability Distributions**: Models similarities as probabilities in high and low dimensions
- **Perplexity**: Controls the effective number of nearest neighbors (typical range: 5-50)
- **Non-linear**: Can capture complex, non-linear relationships
- **Local Structure**: Preserves local neighborhood structure better than global

### Algorithm Steps:
1. **High-D Similarities**: $p_{ij} = \frac{\exp(-||x_i - x_j||^2 / 2\sigma_i^2)}{\sum_{k \neq i} \exp(-||x_i - x_k||^2 / 2\sigma_i^2)}$
2. **Low-D Similarities**: $q_{ij} = \frac{(1 + ||y_i - y_j||^2)^{-1}}{\sum_{k \neq l}(1 + ||y_k - y_l||^2)^{-1}}$
3. **Minimize KL Divergence**: $KL(P||Q) = \sum_i \sum_j p_{ij} \log \frac{p_{ij}}{q_{ij}}$

### Important Notes:
- **Stochastic**: Different runs produce different results
- **Distances**: Distances in t-SNE plot are not meaningful
- **Cluster Sizes**: Cluster sizes may not reflect true proportions

## Key Takeaways

### PCA Advantages:
- **Linear and Fast**: Computationally efficient for large datasets
- **Interpretable**: Components have clear mathematical meaning
- **Deterministic**: Same results every time
- **Global Structure**: Preserves global variance structure
- **Reconstruction**: Can reconstruct original data

### PCA Limitations:
- **Linear Only**: Cannot capture non-linear relationships
- **Assumes Gaussianity**: Works best with normally distributed data
- **Outlier Sensitive**: Affected by outliers in the data

### t-SNE Advantages:
- **Non-linear**: Captures complex, non-linear structures
- **Local Structure**: Excellent for preserving local neighborhoods
- **Visualization**: Outstanding for 2D/3D visualization
- **Cluster Discovery**: Reveals hidden cluster structures

### t-SNE Limitations:
- **Stochastic**: Different results on different runs
- **Slow**: Computationally expensive for large datasets
- **Parameter Sensitive**: Results depend heavily on perplexity
- **Not for ML Pipeline**: Primarily for visualization, not feature extraction
- **Distance Meaningless**: Distances in t-SNE space are not interpretable

### When to Use Which:
- **PCA**: Feature extraction, preprocessing, noise reduction, interpretability needed
- **t-SNE**: Data exploration, visualization, cluster discovery, non-linear structures

## Unsupervised Learning Overview

**Goal**: Find hidden patterns in data without labeled targets.

### Types of Unsupervised Learning
- **Clustering**: Group similar data points
- **Association Rules**: Find relationships between features
- **Dimensionality Reduction**: Reduce feature space
- **Anomaly Detection**: Identify outliers

## k-Means Clustering

**Objective**: Partition data into k clusters by minimizing within-cluster sum of squares (WCSS).

### Algorithm Steps

1. **Initialize**: Choose k random centroids
2. **Assignment**: Assign each point to nearest centroid
3. **Update**: Recalculate centroids as cluster means
4. **Repeat**: Until centroids stabilize or max iterations reached

### Mathematical Formulation

**Objective Function**:
$$J = \sum_{i=1}^{k} \sum_{x \in C_i} ||x - \mu_i||^2$$

Where:
- $C_i$ = cluster $i$
- $\mu_i$ = centroid of cluster $i$
- $||x - \mu_i||^2$ = squared Euclidean distance

### Centroid Update Rule
$$\mu_i = \frac{1}{|C_i|} \sum_{x \in C_i} x$$


## k-Means Variants

### k-Means++
**Improvement**: Smart initialization to avoid poor local optima
- Choose first centroid randomly
- Choose subsequent centroids with probability proportional to squared distance from nearest existing centroid

### Mini-Batch k-Means
**Improvement**: Faster training using random subsets
- Use small random batches for centroid updates
- Trades accuracy for speed on large datasets

### k-Medoids (PAM)
**Improvement**: Uses actual data points as centroids
- More robust to outliers
- Works with any distance metric

## Hierarchical Clustering

**Goal**: Create tree-like cluster structure (dendrogram).

### Types

#### Agglomerative (Bottom-Up)
1. Start with each point as separate cluster
2. Merge closest clusters iteratively
3. Continue until single cluster remains

#### Divisive (Top-Down)
1. Start with all points in one cluster
2. Split clusters recursively
3. Continue until desired number of clusters

### Linkage Criteria


#### Single Linkage
$$d(C_i, C_j) = \min_{x \in C_i, y \in C_j} d(x, y)$$

#### Complete Linkage
$$d(C_i, C_j) = \max_{x \in C_i, y \in C_j} d(x, y)$$

#### Average Linkage
$$d(C_i, C_j) = \frac{1}{|C_i||C_j|} \sum_{x \in C_i} \sum_{y \in C_j} d(x, y)$$

#### Ward Linkage
Minimizes increase in within-cluster variance when merging.


## Choosing Optimal Number of Clusters

### Elbow Method
Plot WCSS vs number of clusters; look for "elbow" point.

### Silhouette Analysis
$$s(i) = \frac{b(i) - a(i)}{\max(a(i), b(i))}$$

Where:
- $a(i)$ = average distance to points in same cluster
- $b(i)$ = average distance to points in nearest cluster
- $s(i) \in [-1, 1]$ (higher is better)

### Gap Statistic
Compare WCSS with expected WCSS from random data.


## Cluster Evaluation Metrics

### Internal Metrics (No ground truth needed)
- **Silhouette Score**: Average silhouette coefficient
- **Calinski-Harabasz Index**: Ratio of between-cluster to within-cluster variance
- **Davies-Bouldin Index**: Average similarity between clusters

### External Metrics (Ground truth available)
- **Adjusted Rand Index (ARI)**: Measures clustering similarity
- **Normalized Mutual Information (NMI)**: Information-theoretic measure
- **Homogeneity & Completeness**: Cluster purity measures


## Applications

### Customer Segmentation
- Group customers by purchasing behavior
- Personalize marketing strategies
- Identify high-value customer segments

### Gene Expression Analysis
- Group genes with similar expression patterns
- Identify functional gene modules
- Drug discovery applications

### Image Segmentation
- Group pixels by color/texture similarity
- Object detection preprocessing
- Medical image analysis

### Market Segmentation
- Group products/services by features
- Competitive analysis
- Strategic positioning


## Advantages & Disadvantages

### k-Means
**Advantages**:
- Fast and simple
- Works well with spherical clusters
- Scales to large datasets

**Disadvantages**:
- Requires specifying k
- Sensitive to initialization
- Assumes spherical clusters
- Sensitive to outliers

### Hierarchical Clustering
**Advantages**:
- No need to specify number of clusters
- Creates interpretable dendrogram
- Deterministic results

**Disadvantages**:
- Computationally expensive O(n³)
- Sensitive to noise and outliers
- Difficult to handle large datasets

## Best Practices

### Data Preprocessing
- **Scale features**: Use StandardScaler or MinMaxScaler
- **Handle outliers**: Remove or transform extreme values
- **Feature selection**: Remove irrelevant features

### Algorithm Selection
- **k-Means**: Large datasets, spherical clusters
- **Hierarchical**: Small datasets, interpretability needed
- **DBSCAN**: Non-spherical clusters, varying densities

### Validation
- **Multiple runs**: k-means with different initializations
- **Cross-validation**: Stability across data subsets
- **Domain knowledge**: Interpret clusters meaningfully

## Key Takeaways
- k-means is fast but requires choosing k and assumes spherical clusters
- Hierarchical clustering provides interpretable dendrograms but is computationally expensive
- Proper data preprocessing is crucial for clustering success
- Combine multiple evaluation metrics for robust cluster assessment

````python
# Clustering Algorithms: k-Means and Hierarchical Clustering Demo
from sklearn.cluster import KMeans, AgglomerativeClustering
from sklearn.datasets import make_blobs, make_circles, make_moons
from sklearn.metrics import silhouette_score, adjusted_rand_score, calinski_harabasz_score
from sklearn.preprocessing import StandardScaler
from scipy.cluster.hierarchy import dendrogram, linkage
from scipy.spatial.distance import pdist

print("=== CLUSTERING ALGORITHMS DEMO ===")

# Demo 1: k-Means on Blob Data
print("\n=== K-MEANS CLUSTERING DEMO ===")

# Generate synthetic blob data
X_blobs, y_true_blobs = make_blobs(n_samples=300, centers=4, n_features=2,
                                  random_state=42, cluster_std=1.5)

print(f"Generated blob data: {X_blobs.shape}")
print(f"True number of clusters: {len(np.unique(y_true_blobs))}")

# Standardize the data
scaler_blobs = StandardScaler()
X_blobs_scaled = scaler_blobs.fit_transform(X_blobs)

# Apply k-means with different k values
k_values = range(2, 8)
kmeans_results = []
inertias = []  # Within-cluster sum of squares (WCSS)
silhouette_scores = []

for k in k_values:
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    cluster_labels = kmeans.fit_predict(X_blobs_scaled)

    # Calculate metrics
    inertia = kmeans.inertia_
    silhouette_avg = silhouette_score(X_blobs_scaled, cluster_labels)

    inertias.append(inertia)
    silhouette_scores.append(silhouette_avg)

    kmeans_results.append({
        'k': k,
        'inertia': inertia,
        'silhouette_score': silhouette_avg,
        'labels': cluster_labels
    })

print("k-Means Results (Elbow Method & Silhouette Analysis):")
for result in kmeans_results:
    print(f"k={result['k']}: Inertia={result['inertia']:.2f}, Silhouette={result['silhouette_score']:.3f}")

# Find optimal k based on silhouette score
optimal_k = max(kmeans_results, key=lambda x: x['silhouette_score'])['k']
print(f"Optimal k based on silhouette score: {optimal_k}")

# Demo 2: Hierarchical Clustering
print("\n=== HIERARCHICAL CLUSTERING DEMO ===")

# Use a smaller dataset for hierarchical clustering (computationally expensive)
X_small = X_blobs_scaled[:100]  # Use subset for demo
y_small = y_true_blobs[:100]

# Apply hierarchical clustering with different linkage methods
linkage_methods = ['ward', 'complete', 'average', 'single']
hierarchical_results = []

for method in linkage_methods:
    hierarchical = AgglomerativeClustering(n_clusters=4, linkage=method)
    cluster_labels = hierarchical.fit_predict(X_small)

    # Calculate silhouette score
    silhouette_avg = silhouette_score(X_small, cluster_labels)
    ari_score = adjusted_rand_score(y_small, cluster_labels)

    hierarchical_results.append({
        'method': method,
        'silhouette_score': silhouette_avg,
        'ari_score': ari_score,
        'labels': cluster_labels
    })

print("Hierarchical Clustering Results:")
for result in hierarchical_results:
    print(f"{result['method']:>8}: Silhouette={result['silhouette_score']:.3f}, ARI={result['ari_score']:.3f}")

# Demo 3: Different Data Shapes
print("\n=== CLUSTERING ON DIFFERENT DATA SHAPES ===")

# Create datasets with different shapes
datasets = {
    'Blobs': make_blobs(n_samples=200, centers=3, n_features=2, random_state=42),
    'Circles': make_circles(n_samples=200, noise=0.1, factor=0.3, random_state=42),
    'Moons': make_moons(n_samples=200, noise=0.1, random_state=42)
}

shape_results = {}

for name, (X, y_true) in datasets.items():
    # Standardize data
    scaler = StandardScaler()
    X_scaled = scaler.fit_transform(X)

    # Apply k-means
    kmeans = KMeans(n_clusters=len(np.unique(y_true)), random_state=42)
    kmeans_labels = kmeans.fit_predict(X_scaled)

    # Apply hierarchical clustering
    hierarchical = AgglomerativeClustering(n_clusters=len(np.unique(y_true)))
    hierarchical_labels = hierarchical.fit_predict(X_scaled)

    # Calculate metrics
    kmeans_sil = silhouette_score(X_scaled, kmeans_labels)
    hierarchical_sil = silhouette_score(X_scaled, hierarchical_labels)
    kmeans_ari = adjusted_rand_score(y_true, kmeans_labels)
    hierarchical_ari = adjusted_rand_score(y_true, hierarchical_labels)

    shape_results[name] = {
        'X': X_scaled,
        'y_true': y_true,
        'kmeans_labels': kmeans_labels,
        'hierarchical_labels': hierarchical_labels,
        'kmeans_sil': kmeans_sil,
        'hierarchical_sil': hierarchical_sil,
        'kmeans_ari': kmeans_ari,
        'hierarchical_ari': hierarchical_ari
    }

print("Performance on Different Data Shapes:")
for name, results in shape_results.items():
    print(f"{name}:")
    print(f"  k-Means:      Silhouette={results['kmeans_sil']:.3f}, ARI={results['kmeans_ari']:.3f}")
    print(f"  Hierarchical: Silhouette={results['hierarchical_sil']:.3f}, ARI={results['hierarchical_ari']:.3f}")

# Demo 4: Cluster Validation Metrics
print("\n=== CLUSTER VALIDATION METRICS DEMO ===")

# Use optimal k-means result for validation
optimal_kmeans = KMeans(n_clusters=optimal_k, random_state=42)
optimal_labels = optimal_kmeans.fit_predict(X_blobs_scaled)

# Calculate various internal metrics
silhouette_avg = silhouette_score(X_blobs_scaled, optimal_labels)
calinski_harabasz = calinski_harabasz_score(X_blobs_scaled, optimal_labels)

# External metric (since we have ground truth)
ari = adjusted_rand_score(y_true_blobs, optimal_labels)

print(f"Validation Metrics for k-Means with k={optimal_k}:")
print(f"Silhouette Score: {silhouette_avg:.3f}")
print(f"Calinski-Harabasz Index: {calinski_harabasz:.2f}")
print(f"Adjusted Rand Index: {ari:.3f}")

# Visualization
plt.figure(figsize=(18, 12))

# Plot 1: Original data
plt.subplot(3, 4, 1)
colors = ['red', 'blue', 'green', 'purple']
for i, color in enumerate(colors):
    mask = y_true_blobs == i
    plt.scatter(X_blobs[mask, 0], X_blobs[mask, 1], c=color, alpha=0.6, s=50)
plt.title('Original Data (Ground Truth)')
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.grid(True, alpha=0.3)

# Plot 2: Elbow method
plt.subplot(3, 4, 2)
plt.plot(k_values, inertias, 'bo-', linewidth=2, markersize=8)
plt.xlabel('Number of Clusters (k)')
plt.ylabel('Inertia (WCSS)')
plt.title('Elbow Method')
plt.grid(True, alpha=0.3)

# Plot 3: Silhouette analysis
plt.subplot(3, 4, 3)
plt.plot(k_values, silhouette_scores, 'ro-', linewidth=2, markersize=8)
plt.axvline(x=optimal_k, color='green', linestyle='--', alpha=0.7, label=f'Optimal k={optimal_k}')
plt.xlabel('Number of Clusters (k)')
plt.ylabel('Average Silhouette Score')
plt.title('Silhouette Analysis')
plt.legend()
plt.grid(True, alpha=0.3)

# Plot 4: Optimal k-means result
plt.subplot(3, 4, 4)
optimal_result = next(r for r in kmeans_results if r['k'] == optimal_k)
labels = optimal_result['labels']
for i, color in enumerate(colors):
    mask = labels == i
    plt.scatter(X_blobs[mask, 0], X_blobs[mask, 1], c=color, alpha=0.6, s=50)
plt.scatter(optimal_kmeans.cluster_centers_[:, 0] * scaler_blobs.scale_[0] + scaler_blobs.mean_[0],
           optimal_kmeans.cluster_centers_[:, 1] * scaler_blobs.scale_[1] + scaler_blobs.mean_[1],
           c='black', marker='x', s=200, linewidths=3)
plt.title(f'k-Means Result (k={optimal_k})')
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.grid(True, alpha=0.3)

# Plots 5-8: Hierarchical clustering with different linkage methods
for idx, result in enumerate(hierarchical_results):
    plt.subplot(3, 4, 5 + idx)
    labels = result['labels']
    for i, color in enumerate(colors):
        mask = labels == i
        plt.scatter(X_small[mask, 0] * scaler_blobs.scale_[0] + scaler_blobs.mean_[0],
                   X_small[mask, 1] * scaler_blobs.scale_[1] + scaler_blobs.mean_[1],
                   c=color, alpha=0.6, s=50)
    plt.title(f'{result["method"].title()} Linkage\nSil: {result["silhouette_score"]:.2f}')
    plt.xlabel('Feature 1')
    plt.ylabel('Feature 2')
    plt.grid(True, alpha=0.3)

# Plots 9-11: Different data shapes
for idx, (name, results) in enumerate(shape_results.items()):
    plt.subplot(3, 4, 9 + idx)

    X = results['X']
    y_true = results['y_true']

    # Show k-means results
    kmeans_labels = results['kmeans_labels']
    for i in range(len(np.unique(kmeans_labels))):
        mask = kmeans_labels == i
        plt.scatter(X[mask, 0], X[mask, 1], alpha=0.6, s=30)

    plt.title(f'{name} - k-Means\nARI: {results["kmeans_ari"]:.2f}')
    plt.xlabel('Feature 1')
    plt.ylabel('Feature 2')
    plt.grid(True, alpha=0.3)

# Plot 12: Dendrogram for hierarchical clustering
plt.subplot(3, 4, 12)
# Create linkage matrix for dendrogram
linkage_matrix = linkage(X_small, method='ward')
dendrogram(linkage_matrix, truncate_mode='lastp', p=10, leaf_rotation=90, leaf_font_size=10)
plt.title('Hierarchical Clustering\nDendrogram (Ward)')
plt.xlabel('Sample Index or (Cluster Size)')
plt.ylabel('Distance')

plt.tight_layout()
plt.show()

print("\n=== CLUSTERING SUMMARY ===")
print("Key Findings:")

# Find best performing method for each dataset
print("Best methods by data shape:")
for name, results in shape_results.items():
    if results['kmeans_ari'] > results['hierarchical_ari']:
        best_method = 'k-Means'
        best_score = results['kmeans_ari']
    else:
        best_method = 'Hierarchical'
        best_score = results['hierarchical_ari']
    print(f"• {name:>8}: {best_method} (ARI: {best_score:.3f})")

print(f"\nOptimal parameters:")
print(f"• k-Means optimal k: {optimal_k} (Silhouette: {silhouette_scores[optimal_k-2]:.3f})")

best_hierarchical = max(hierarchical_results, key=lambda x: x['silhouette_score'])
print(f"• Best hierarchical linkage: {best_hierarchical['method']} (Silhouette: {best_hierarchical['silhouette_score']:.3f})")

print("\nAlgorithm Characteristics:")
print("k-Means:")
print("  • Fast and scalable")
print("  • Requires choosing k in advance")
print("  • Works best with spherical, well-separated clusters")
print("  • Sensitive to initialization")

print("Hierarchical Clustering:")
print("  • No need to specify number of clusters")
print("  • Creates interpretable dendrogram")
print("  • Computationally expensive O(n³)")
print("  • Sensitive to outliers")

print("\nBest Practices:")
print("• Always standardize features before clustering")
print("• Use multiple evaluation metrics")
print("• Try different algorithms and parameters")
print("• Validate results with domain knowledge")
print("• Consider DBSCAN for non-spherical clusters")
````

## Classification Metrics

### Confusion Matrix Foundation
$$\begin{pmatrix}
TN & FP \\
FN & TP
\end{pmatrix}$$

Where:
- **TP**: True Positives (correctly predicted positive)
- **TN**: True Negatives (correctly predicted negative)  
- **FP**: False Positives (incorrectly predicted positive, Type I error)
- **FN**: False Negatives (incorrectly predicted negative, Type II error)


### Basic Classification Metrics

#### Accuracy
$$\text{Accuracy} = \frac{TP + TN}{TP + TN + FP + FN}$$

#### Precision (Positive Predictive Value)
$$\text{Precision} = \frac{TP}{TP + FP}$$

#### Recall (Sensitivity, True Positive Rate)
$$\text{Recall} = \frac{TP}{TP + FN}$$

#### Specificity (True Negative Rate)
$$\text{Specificity} = \frac{TN}{TN + FP}$$

#### F1-Score (Harmonic Mean of Precision and Recall)
$$\text{F1} = 2 \cdot \frac{\text{Precision} \cdot \text{Recall}}{\text{Precision} + \text{Recall}} = \frac{2TP}{2TP + FP + FN}$$


### Advanced Classification Metrics


#### ROC Curve & AUC
- **ROC**: Receiver Operating Characteristic
- **AUC**: Area Under the Curve
- **TPR vs FPR**: True Positive Rate vs False Positive Rate
- **AUC = 1.0**: Perfect classifier
- **AUC = 0.5**: Random classifier

#### Precision-Recall Curve
Better for imbalanced datasets than ROC-AUC
- **High Precision, High Recall**: Ideal
- **Area Under PR Curve**: Summary metric

#### Log Loss (Cross-Entropy Loss)
$$\text{Log Loss} = -\frac{1}{N}\sum_{i=1}^{N}[y_i \log(p_i) + (1-y_i)\log(1-p_i)]$$

#### Cohen's Kappa
Measures inter-rater agreement, accounting for chance:
$$\kappa = \frac{p_o - p_e}{1 - p_e}$$

Where $p_o$ is observed agreement and $p_e$ is expected agreement by chance.


## Regression Metrics

### Mean Absolute Error (MAE)
$$\text{MAE} = \frac{1}{n}\sum_{i=1}^{n}|y_i - \hat{y_i}|$$
- **Robust** to outliers
- **Linear penalty** for errors

### Mean Squared Error (MSE)
$$\text{MSE} = \frac{1}{n}\sum_{i=1}^{n}(y_i - \hat{y_i})^2$$
- **Penalizes large errors** more heavily
- **Same units** as target variable squared

### Root Mean Squared Error (RMSE)
$$\text{RMSE} = \sqrt{\text{MSE}} = \sqrt{\frac{1}{n}\sum_{i=1}^{n}(y_i - \hat{y_i})^2}$$
- **Same units** as target variable
- **More interpretable** than MSE

### R-squared (Coefficient of Determination)
$$R^2 = 1 - \frac{SS_{res}}{SS_{tot}} = 1 - \frac{\sum(y_i - \hat{y_i})^2}{\sum(y_i - \bar{y})^2}$$
- **Proportion of variance** explained by model
- **0 ≤ R² ≤ 1** for models with intercept

### Mean Absolute Percentage Error (MAPE)
$$\text{MAPE} = \frac{100\%}{n}\sum_{i=1}^{n}\left|\frac{y_i - \hat{y_i}}{y_i}\right|$$
- **Scale-independent**
- **Interpretable as percentage**

## Model Selection Considerations

### Bias-Variance Tradeoff

#### Bias
Expected difference between true value and predictions:
$$\text{Bias}[\hat{f}(x)] = E[\hat{f}(x)] - f(x)$$

#### Variance  
Expected squared deviation from expected prediction:
$$\text{Variance}[\hat{f}(x)] = E[(\hat{f}(x) - E[\hat{f}(x)])^2]$$

#### Total Error Decomposition
$$E[(y - \hat{f}(x))^2] = \text{Bias}^2 + \text{Variance} + \text{Irreducible Error}$$

### Overfitting vs Underfitting

#### Overfitting (High Variance)
- **Training error << Validation error**
- **Model too complex** for available data
- **Solutions**: Regularization, more data, simpler model

#### Underfitting (High Bias)
- **Both training and validation error high**
- **Model too simple** to capture patterns
- **Solutions**: More complex model, feature engineering

## Evaluation Strategies

### Train/Validation/Test Split
- **Training Set (60%)**: Model training
- **Validation Set (20%)**: Model selection and tuning
- **Test Set (20%)**: Final evaluation

### Time Series Considerations
- **No random splitting**: Use temporal order
- **Walk-forward validation**: Expanding or sliding window
- **Avoid data leakage**: Future data in past predictions

### Imbalanced Data Evaluation

#### Problems with Accuracy
- **Misleading with skewed classes**
- **Example**: 95% accuracy on 95% majority class

#### Better Metrics for Imbalanced Data
- **Precision and Recall**
- **F1-Score** or F-beta scores
- **ROC-AUC** (can be misleading)
- **PR-AUC** (better for imbalanced data)
- **Balanced Accuracy**: Average of TPR and TNR

## Common Evaluation Pitfalls

### Data Leakage
- **Temporal leakage**: Future information in features
- **Feature leakage**: Features contain target information
- **Preprocessing leakage**: Scaling on entire dataset before splitting

### Multiple Hypothesis Testing
- **Problem**: Testing many models increases false discoveries
- **Solutions**: Bonferroni correction, held-out test set

### Cherry-Picking Metrics
- **Problem**: Selecting metrics that favor your model
- **Solution**: Define evaluation criteria beforehand

### Sample Size Considerations
- **Small datasets**: High variance in metrics
- **Bootstrap confidence intervals** for robust estimates


## Metric Selection Guidelines

### Classification Tasks
- **Balanced classes**: Accuracy, F1-Score
- **Imbalanced classes**: Precision, Recall, PR-AUC
- **Ranking problems**: ROC-AUC
- **Probability calibration**: Log-loss

### Regression Tasks
- **General purpose**: RMSE, MAE
- **Outlier robustness**: MAE
- **Interpretability**: R², MAPE
- **Different scales**: MAPE, normalized metrics


## Business Metrics Integration

### Cost-Sensitive Evaluation
- **False positives cost ≠ False negatives cost**
- **Custom loss functions** based on business impact
- **ROI-based model evaluation**

### Threshold Optimization
- **Business-driven thresholds**
- **Precision-Recall tradeoffs**
- **Cost-benefit analysis**

## Key Takeaways
- Choose metrics aligned with business objectives
- Use multiple complementary metrics for comprehensive evaluation
- Be aware of class imbalance and temporal dependencies
- Validate evaluation methodology to avoid data leakage
- Consider bias-variance tradeoff in model selection

## Cross-Validation Fundamentals

**Purpose**: Obtain more robust performance estimates by using all data for both training and validation.

### K-Fold Cross-Validation
1. **Split** data into k equal-sized folds
2. **Train** on k-1 folds, validate on remaining fold
3. **Repeat** k times, each fold serves as validation once
4. **Average** performance across all folds

$$CV_k = \frac{1}{k}\sum_{i=1}^{k} L(f_i, D_i^{val})$$

Where $f_i$ is model trained on all folds except $i$, $D_i^{val}$ is validation fold $i$.

## Cross-Validation Strategies

### Stratified K-Fold
**Use Case**: Classification with imbalanced classes
**Method**: Maintains class proportions in each fold
**Advantage**: More representative splits for skewed datasets

### Leave-One-Out CV (LOOCV)
**Method**: k = n (number of samples)
**Advantage**: Maximum use of training data
**Disadvantage**: Computationally expensive, high variance estimates

### Time Series CV
**Use Case**: Temporal data where order matters
**Method**: Train on past, validate on future (expanding/sliding window)
**Key**: Respects temporal ordering to avoid data leakage

### Group K-Fold
**Use Case**: Data with natural groupings
**Method**: Entire groups in same fold (prevent data leakage)
**Example**: Patient data where multiple samples per patient

## Hyperparameter Tuning Methods

### Grid Search
**Method**: Exhaustive search over predefined parameter grid
**Advantages**: Systematic, finds global optimum within grid
**Disadvantages**: Exponential complexity, curse of dimensionality

#### Example Grid
```python
param_grid = {
    'n_estimators': [50, 100, 200],
    'max_depth': [3, 5, 7, 10],
    'min_samples_split': [2, 5, 10]
}
# Total combinations: 3 × 4 × 3 = 36
```

### Random Search

**Method**: Random sampling from parameter distributions

**Advantages**: More efficient for high dimensions, can find better solutions

**Theory**: Often 95% of performance achieved with <5% of evaluations


#### Mathematical Insight
For grid search with n parameters, trying k values per parameter requires $k^n$ evaluations.
Random search tries random combinations, more likely to hit good values in high dimensions.

### Bayesian Optimization
**Method**: Use probabilistic model to guide search
**Key Components**:
- **Surrogate Model**: Gaussian Process modeling objective function
- **Acquisition Function**: Balances exploration vs exploitation

#### Popular Acquisition Functions
- **Expected Improvement (EI)**
- **Upper Confidence Bound (UCB)**
- **Probability of Improvement (PI)**

### Successive Halving
**Method**: Start with many configurations, progressively eliminate poor performers
**Advantage**: Efficient for expensive evaluations
**Implementation**: Hyperband, BOHB algorithms


## Advanced Validation Techniques

### Nested Cross-Validation
**Purpose**: Unbiased model selection and performance estimation
**Structure**:
- **Outer Loop**: Performance estimation
- **Inner Loop**: Hyperparameter tuning

```
For each outer fold:
    Use remaining data for inner CV to tune hyperparameters
    Train final model with best hyperparameters
    Evaluate on held-out outer fold
```

### Cross-Validation with Preprocessing
**Critical**: Apply preprocessing within each CV fold
**Wrong Way**: Preprocess entire dataset, then CV
**Right Way**: Preprocess within each fold separately

```python
# Correct approach
for train_idx, val_idx in cv_folds:
    X_train, X_val = X[train_idx], X[val_idx]
    
    # Fit preprocessing on training data only
    scaler = StandardScaler().fit(X_train)
    X_train_scaled = scaler.transform(X_train)
    X_val_scaled = scaler.transform(X_val)
    
    # Train and evaluate model
```


## Hyperparameter Categories

### Model-Specific Parameters

#### Tree-Based Models
- **max_depth**: Maximum tree depth
- **min_samples_split**: Minimum samples to split node
- **min_samples_leaf**: Minimum samples in leaf
- **max_features**: Features considered per split

#### Neural Networks
- **learning_rate**: Step size for optimization
- **batch_size**: Samples per gradient update
- **hidden_units**: Neurons per layer
- **dropout_rate**: Regularization probability

#### SVM
- **C**: Regularization strength
- **gamma**: RBF kernel parameter
- **kernel**: Kernel type (linear, rbf, poly)


### Regularization Parameters
- **L1/L2 regularization**: α, λ parameters
- **Elastic Net**: Balance between L1 and L2
- **Dropout**: Neural network regularization

### Optimization Parameters
- **Solver**: Optimization algorithm
- **max_iterations**: Convergence limit
- **tolerance**: Convergence threshold

## Validation Strategies by Problem Type

### Small Datasets (n < 1000)
- **LOOCV** or high k (k=10+)
- **Stratified CV** for classification
- **Bootstrap validation** for very small datasets

### Large Datasets (n > 100,000)
- **Lower k** (k=3-5) for efficiency
- **Single holdout** validation set
- **Temporal splits** for time series

### Time Series
- **Walk-forward validation**
- **Expanding window**: Growing training set
- **Sliding window**: Fixed-size training set

### Imbalanced Data
- **Stratified CV** mandatory
- **Appropriate metrics**: F1, AUC-PR, not accuracy
- **Consider cost-sensitive evaluation**

## Early Stopping & Validation

### Early Stopping
**Method**: Stop training when validation performance stops improving
**Benefits**: Prevents overfitting, reduces training time
**Implementation**: Monitor validation loss with patience parameter

### Validation Curves
**Purpose**: Understand hyperparameter impact
**Method**: Plot performance vs parameter values
**Insight**: Identify underfitting/overfitting regions

### Learning Curves
**Purpose**: Diagnose model performance vs dataset size
**Insights**:
- **High bias**: Both curves plateau at low performance
- **High variance**: Large gap between training and validation

## Best Practices

### Computational Efficiency
- **Start simple**: Begin with default parameters
- **Coarse-to-fine**: Broad search first, then refine
- **Early stopping**: Use validation-based stopping
- **Parallel processing**: Leverage multiple cores

### Statistical Significance
- **Multiple seeds**: Run with different random seeds
- **Confidence intervals**: Report uncertainty in estimates
- **Statistical tests**: Compare models statistically

### Avoiding Overfitting
- **Separate test set**: Never use for hyperparameter tuning
- **Nested CV**: Unbiased performance estimation
- **Domain knowledge**: Use reasonable parameter ranges

## Common Pitfalls

### Data Leakage in CV
- **Preprocessing leakage**: Scaling/encoding before CV
- **Temporal leakage**: Future information in past predictions
- **Group leakage**: Related samples in different folds

### Overfitting to Validation Set
- **Problem**: Extensive hyperparameter tuning on same validation set
- **Solution**: Nested CV or separate test set

### Inappropriate CV Strategy
- **Time series**: Using random CV instead of temporal splits
- **Grouped data**: Ignoring natural groupings


## Model Selection Criteria

### Performance Metrics
- **Primary metric**: Aligned with business objective
- **Secondary metrics**: Additional perspectives
- **Computational cost**: Training and inference time

### Robustness
- **Stability**: Consistent performance across CV folds
- **Generalization**: Performance on diverse validation sets
- **Interpretability**: Model complexity vs performance tradeoff

## Key Takeaways
- Cross-validation provides robust performance estimates
- Hyperparameter tuning requires systematic approach
- Nested CV gives unbiased model selection
- Avoid data leakage in preprocessing and validation
- Choose CV strategy appropriate for your data characteristics

````python
# Cross-Validation & Hyperparameter Tuning Comprehensive Demo
from sklearn.model_selection import (
    cross_val_score, StratifiedKFold, KFold, LeaveOneOut,
    GridSearchCV, RandomizedSearchCV, validation_curve,
    TimeSeriesSplit, GroupKFold
)
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC
from sklearn.linear_model import LogisticRegression
from sklearn.datasets import make_classification, make_regression
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
from scipy.stats import randint, uniform
import time

print("=== CROSS-VALIDATION & HYPERPARAMETER TUNING DEMO ===")

# Generate dataset for demonstrations
X_cv, y_cv = make_classification(
    n_samples=1000, n_features=20, n_informative=15, n_redundant=5,
    n_classes=3, random_state=42
)

print(f"Dataset shape: {X_cv.shape}")
print(f"Classes distribution: {np.bincount(y_cv)}")

# Demo 1: Different Cross-Validation Strategies
print("\n=== CROSS-VALIDATION STRATEGIES COMPARISON ===")

# Create models for comparison
models = {
    'Logistic Regression': LogisticRegression(random_state=42, max_iter=1000),
    'Random Forest': RandomForestClassifier(n_estimators=50, random_state=42),
    'SVM': SVC(random_state=42)
}

# Different CV strategies
cv_strategies = {
    'K-Fold (k=5)': KFold(n_splits=5, shuffle=True, random_state=42),
    'Stratified K-Fold (k=5)': StratifiedKFold(n_splits=5, shuffle=True, random_state=42),
    'K-Fold (k=10)': KFold(n_splits=10, shuffle=True, random_state=42),
    'Leave-One-Out': LeaveOneOut() if len(X_cv) <= 200 else StratifiedKFold(n_splits=min(len(X_cv)//10, 20), shuffle=True, random_state=42)
}

# Note: Using subset for LOOCV due to computational constraints
if len(X_cv) > 200:
    print("Note: Using subset for LOOCV demonstration due to computational constraints")
    X_loocv, y_loocv = X_cv[:200], y_cv[:200]
    cv_strategies['Leave-One-Out'] = LeaveOneOut()
else:
    X_loocv, y_loocv = X_cv, y_cv

cv_results = {}

for model_name, model in models.items():
    cv_results[model_name] = {}
    print(f"\n{model_name} Results:")

    for cv_name, cv_strategy in cv_strategies.items():
        if cv_name == 'Leave-One-Out':
            # Use smaller dataset for LOOCV
            scores = cross_val_score(model, X_loocv, y_loocv, cv=cv_strategy, scoring='accuracy')
        else:
            scores = cross_val_score(model, X_cv, y_cv, cv=cv_strategy, scoring='accuracy')

        cv_results[model_name][cv_name] = {
            'mean': scores.mean(),
            'std': scores.std(),
            'scores': scores
        }

        print(f"  {cv_name:>20}: {scores.mean():.3f} ± {scores.std():.3f}")

# Demo 2: Grid Search Hyperparameter Tuning
print("\n=== GRID SEARCH HYPERPARAMETER TUNING ===")

# Define parameter grids
param_grids = {
    'Random Forest': {
        'n_estimators': [50, 100, 200],
        'max_depth': [3, 5, 7, None],
        'min_samples_split': [2, 5, 10],
        'min_samples_leaf': [1, 2, 4]
    },
    'SVM': {
        'C': [0.1, 1, 10, 100],
        'gamma': ['scale', 'auto', 0.001, 0.01, 0.1, 1],
        'kernel': ['rbf', 'linear']
    },
    'Logistic Regression': {
        'C': [0.01, 0.1, 1, 10, 100],
        'penalty': ['l1', 'l2'],
        'solver': ['liblinear']  # Supports both l1 and l2
    }
}

grid_search_results = {}

for model_name in ['Random Forest', 'SVM']:  # Skip LogReg for time
    print(f"\nGrid Search for {model_name}:")

    if model_name == 'Random Forest':
        base_model = RandomForestClassifier(random_state=42)
        param_grid = param_grids[model_name]
        # Reduce grid size for demo
        param_grid = {
            'n_estimators': [50, 100],
            'max_depth': [3, 5, None],
            'min_samples_split': [2, 5]
        }
    else:  # SVM
        base_model = SVC(random_state=42)
        param_grid = param_grids[model_name]
        # Reduce grid size for demo
        param_grid = {
            'C': [0.1, 1, 10],
            'gamma': ['scale', 0.01, 0.1],
            'kernel': ['rbf', 'linear']
        }

    start_time = time.time()
    grid_search = GridSearchCV(
        base_model, param_grid, cv=5, scoring='accuracy',
        n_jobs=-1, return_train_score=True
    )
    grid_search.fit(X_cv, y_cv)
    end_time = time.time()

    grid_search_results[model_name] = {
        'best_params': grid_search.best_params_,
        'best_score': grid_search.best_score_,
        'time_taken': end_time - start_time,
        'n_combinations': len(grid_search.cv_results_['mean_test_score'])
    }

    print(f"  Best parameters: {grid_search.best_params_}")
    print(f"  Best CV score: {grid_search.best_score_:.3f}")
    print(f"  Total combinations tested: {len(grid_search.cv_results_['mean_test_score'])}")
    print(f"  Time taken: {end_time - start_time:.2f} seconds")

# Demo 3: Random Search vs Grid Search Comparison
print("\n=== RANDOM SEARCH VS GRID SEARCH COMPARISON ===")

# Focus on Random Forest for detailed comparison
rf_param_dist = {
    'n_estimators': randint(10, 200),
    'max_depth': [3, 5, 7, 10, None],
    'min_samples_split': randint(2, 20),
    'min_samples_leaf': randint(1, 10),
    'max_features': ['sqrt', 'log2', None]
}

# Grid search (smaller grid)
rf_param_grid = {
    'n_estimators': [50, 100, 150],
    'max_depth': [5, 7, None],
    'min_samples_split': [2, 5, 10]
}

# Random search
print("Random Search:")
start_time = time.time()
random_search = RandomizedSearchCV(
    RandomForestClassifier(random_state=42), rf_param_dist,
    n_iter=20, cv=5, scoring='accuracy', n_jobs=-1, random_state=42
)
random_search.fit(X_cv, y_cv)
random_time = time.time() - start_time

print(f"  Best parameters: {random_search.best_params_}")
print(f"  Best CV score: {random_search.best_score_:.3f}")
print(f"  Time taken: {random_time:.2f} seconds")
print(f"  Combinations tested: 20")

# Grid search
print("\nGrid Search (reduced grid):")
start_time = time.time()
grid_search_rf = GridSearchCV(
    RandomForestClassifier(random_state=42), rf_param_grid,
    cv=5, scoring='accuracy', n_jobs=-1
)
grid_search_rf.fit(X_cv, y_cv)
grid_time = time.time() - start_time

print(f"  Best parameters: {grid_search_rf.best_params_}")
print(f"  Best CV score: {grid_search_rf.best_score_:.3f}")
print(f"  Time taken: {grid_time:.2f} seconds")
print(f"  Combinations tested: {len(grid_search_rf.cv_results_['mean_test_score'])}")

# Demo 4: Nested Cross-Validation
print("\n=== NESTED CROSS-VALIDATION DEMO ===")

def nested_cv_score(model, param_grid, X, y, outer_cv=5, inner_cv=3):
    """
    Perform nested cross-validation for unbiased model evaluation
    """
    outer_scores = []
    outer_cv_splitter = StratifiedKFold(n_splits=outer_cv, shuffle=True, random_state=42)

    for train_idx, test_idx in outer_cv_splitter.split(X, y):
        X_train_outer, X_test_outer = X[train_idx], X[test_idx]
        y_train_outer, y_test_outer = y[train_idx], y[test_idx]

        # Inner CV for hyperparameter tuning
        inner_cv_splitter = StratifiedKFold(n_splits=inner_cv, shuffle=True, random_state=42)
        grid_search = GridSearchCV(
            model, param_grid, cv=inner_cv_splitter,
            scoring='accuracy', n_jobs=-1
        )
        grid_search.fit(X_train_outer, y_train_outer)

        # Evaluate best model on outer test set
        best_model = grid_search.best_estimator_
        outer_score = best_model.score(X_test_outer, y_test_outer)
        outer_scores.append(outer_score)

    return np.array(outer_scores)

# Perform nested CV on Random Forest
nested_param_grid = {
    'n_estimators': [50, 100],
    'max_depth': [3, 5, None]
}

print("Performing Nested Cross-Validation (may take a moment)...")
nested_scores = nested_cv_score(
    RandomForestClassifier(random_state=42),
    nested_param_grid, X_cv, y_cv
)

print(f"Nested CV Scores: {nested_scores}")
print(f"Mean nested CV score: {nested_scores.mean():.3f} ± {nested_scores.std():.3f}")
print(f"This represents unbiased estimate of model performance")

# Demo 5: Validation Curves
print("\n=== VALIDATION CURVES ANALYSIS ===")

# Validation curve for Random Forest n_estimators
param_range = [10, 25, 50, 100, 150, 200]
train_scores, val_scores = validation_curve(
    RandomForestClassifier(random_state=42), X_cv, y_cv,
    param_name='n_estimators', param_range=param_range,
    cv=5, scoring='accuracy', n_jobs=-1
)

train_mean = np.mean(train_scores, axis=1)
train_std = np.std(train_scores, axis=1)
val_mean = np.mean(val_scores, axis=1)
val_std = np.std(val_scores, axis=1)

print("Validation Curve Results (n_estimators):")
for i, n_est in enumerate(param_range):
    print(f"  n_estimators={n_est:3d}: Train={train_mean[i]:.3f}±{train_std[i]:.3f}, "
          f"Val={val_mean[i]:.3f}±{val_std[i]:.3f}")

# Demo 6: Time Series Cross-Validation
print("\n=== TIME SERIES CROSS-VALIDATION DEMO ===")

# Generate time series-like data
np.random.seed(42)
n_samples = 200
X_ts = np.random.randn(n_samples, 5)
# Add time dependency
for i in range(1, n_samples):
    X_ts[i, 0] += 0.5 * X_ts[i-1, 0]  # AR(1) process
y_ts = (X_ts[:, 0] + X_ts[:, 1] + np.random.randn(n_samples) * 0.1 > 0).astype(int)

print(f"Time series dataset shape: {X_ts.shape}")

# Compare regular CV vs Time Series CV
regular_cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
ts_cv = TimeSeriesSplit(n_splits=5)

model_ts = RandomForestClassifier(n_estimators=50, random_state=42)

# Regular CV (wrong for time series!)
regular_scores = cross_val_score(model_ts, X_ts, y_ts, cv=regular_cv)
print(f"Regular K-Fold CV (inappropriate): {regular_scores.mean():.3f} ± {regular_scores.std():.3f}")

# Time Series CV (correct approach)
ts_scores = cross_val_score(model_ts, X_ts, y_ts, cv=ts_cv)
print(f"Time Series CV (appropriate): {ts_scores.mean():.3f} ± {ts_scores.std():.3f}")

print("Note: Time Series CV typically gives more conservative estimates")

# Visualization
plt.figure(figsize=(18, 12))

# Plot 1: CV Strategy Comparison
plt.subplot(2, 3, 1)
cv_methods = list(cv_strategies.keys())[:3]  # First 3 methods
lr_scores = [cv_results['Logistic Regression'][method]['mean'] for method in cv_methods]
rf_scores = [cv_results['Random Forest'][method]['mean'] for method in cv_methods]
svm_scores = [cv_results['SVM'][method]['mean'] for method in cv_methods]

x = np.arange(len(cv_methods))
width = 0.25

plt.bar(x - width, lr_scores, width, label='Logistic Regression', alpha=0.8)
plt.bar(x, rf_scores, width, label='Random Forest', alpha=0.8)
plt.bar(x + width, svm_scores, width, label='SVM', alpha=0.8)

plt.xlabel('CV Strategy')
plt.ylabel('Accuracy')
plt.title('CV Strategy Comparison')
plt.xticks(x, cv_methods, rotation=45)
plt.legend()
plt.grid(True, alpha=0.3)

# Plot 2: Grid Search vs Random Search
plt.subplot(2, 3, 2)
methods = ['Random Search\n(20 iter)', 'Grid Search\n(27 combinations)']
scores = [random_search.best_score_, grid_search_rf.best_score_]
times = [random_time, grid_time]

fig_ax = plt.gca()
color = 'tab:blue'
fig_ax.set_xlabel('Search Method')
fig_ax.set_ylabel('Best CV Score', color=color)
bars1 = fig_ax.bar(methods, scores, color=color, alpha=0.7)
fig_ax.tick_params(axis='y', labelcolor=color)

# Add time on secondary y-axis
ax2 = fig_ax.twinx()
color = 'tab:red'
ax2.set_ylabel('Time (seconds)', color=color)
bars2 = ax2.bar(methods, times, color=color, alpha=0.5, width=0.6)
ax2.tick_params(axis='y', labelcolor=color)

plt.title('Grid Search vs Random Search')

# Add value labels
for bar, score in zip(bars1, scores):
    plt.text(bar.get_x() + bar.get_width()/2, bar.get_height() + 0.005,
             f'{score:.3f}', ha='center', va='bottom', color='blue')

# Plot 3: Validation Curve
plt.subplot(2, 3, 3)
plt.plot(param_range, train_mean, 'o-', label='Training Score', color='blue')
plt.fill_between(param_range, train_mean - train_std, train_mean + train_std, alpha=0.1, color='blue')
plt.plot(param_range, val_mean, 'o-', label='Validation Score', color='red')
plt.fill_between(param_range, val_mean - val_std, val_mean + val_std, alpha=0.1, color='red')

plt.xlabel('n_estimators')
plt.ylabel('Accuracy')
plt.title('Validation Curve: n_estimators')
plt.legend()
plt.grid(True, alpha=0.3)

# Plot 4: Nested CV Visualization
plt.subplot(2, 3, 4)
fold_indices = np.arange(1, len(nested_scores) + 1)
bars = plt.bar(fold_indices, nested_scores, alpha=0.7, color='green')
plt.axhline(y=nested_scores.mean(), color='red', linestyle='--',
           label=f'Mean: {nested_scores.mean():.3f}')

plt.xlabel('Outer CV Fold')
plt.ylabel('Accuracy')
plt.title('Nested CV Results')
plt.legend()
plt.grid(True, alpha=0.3)

# Add value labels
for bar, score in zip(bars, nested_scores):
    plt.text(bar.get_x() + bar.get_width()/2, bar.get_height() + 0.01,
             f'{score:.3f}', ha='center', va='bottom')

# Plot 5: Time Series CV Illustration
plt.subplot(2, 3, 5)
comparison_data = ['Regular CV\n(Wrong)', 'Time Series CV\n(Correct)']
ts_comparison_scores = [regular_scores.mean(), ts_scores.mean()]
colors = ['red', 'green']

bars = plt.bar(comparison_data, ts_comparison_scores, color=colors, alpha=0.7)
plt.ylabel('Accuracy')
plt.title('Time Series CV vs Regular CV')
plt.grid(True, alpha=0.3)

# Add value labels and error bars
plt.errorbar(comparison_data, ts_comparison_scores,
            yerr=[regular_scores.std(), ts_scores.std()],
            fmt='o', color='black', capsize=5)

for bar, score in zip(bars, ts_comparison_scores):
    plt.text(bar.get_x() + bar.get_width()/2, bar.get_height() + 0.01,
             f'{score:.3f}', ha='center', va='bottom')

# Plot 6: Hyperparameter Tuning Summary
plt.subplot(2, 3, 6)
plt.axis('off')

summary_text = f"""
HYPERPARAMETER TUNING RESULTS

Grid Search Results:
• Random Forest: {grid_search_results['Random Forest']['best_score']:.3f}
  Time: {grid_search_results['Random Forest']['time_taken']:.1f}s
• SVM: {grid_search_results['SVM']['best_score']:.3f}
  Time: {grid_search_results['SVM']['time_taken']:.1f}s

Random Search vs Grid Search:
• Random (20 iter): {random_search.best_score_:.3f} ({random_time:.1f}s)
• Grid (27 combos): {grid_search_rf.best_score_:.3f} ({grid_time:.1f}s)

Nested CV (Unbiased): {nested_scores.mean():.3f} ± {nested_scores.std():.3f}

Time Series Insight:
• Regular CV overestimates performance
• Always use temporal splits for time series
"""

plt.text(0.05, 0.95, summary_text, transform=plt.gca().transAxes,
         fontsize=10, verticalalignment='top', fontfamily='monospace',
         bbox=dict(boxstyle='round', facecolor='lightgray', alpha=0.8))

plt.tight_layout()
plt.show()

print("\n=== CROSS-VALIDATION & HYPERPARAMETER TUNING SUMMARY ===")
print("Key Insights:")

# Best CV strategy
best_cv_method = max(cv_strategies.keys(),
                    key=lambda x: cv_results['Random Forest'][x]['mean']
                    if x != 'Leave-One-Out' else 0)
print(f"• Most stable CV method: {best_cv_method}")

# Search method comparison
if random_search.best_score_ > grid_search_rf.best_score_:
    better_search = "Random Search"
    worse_search = "Grid Search"
else:
    better_search = "Grid Search"
    worse_search = "Random Search"

print(f"• {better_search} found better hyperparameters than {worse_search}")
print(f"• Random search was {grid_time/random_time:.1f}x faster than grid search")

print(f"• Nested CV provides unbiased estimate: {nested_scores.mean():.3f} ± {nested_scores.std():.3f}")
print(f"• Time series CV is more conservative: {ts_scores.mean():.3f} vs {regular_scores.mean():.3f}")

print("\nBest Practices Demonstrated:")
print("• Use stratified CV for classification")
print("• Random search often more efficient than grid search")
print("• Nested CV provides unbiased model selection")
print("• Time series requires temporal CV splits")
print("• Validation curves help identify optimal hyperparameters")
print("• Always separate hyperparameter tuning from final evaluation")

print("\nCritical Reminders:")
print("• Never use test set for hyperparameter tuning")
print("• Apply preprocessing within each CV fold")
print("• Choose CV strategy based on data characteristics")
print("• Consider computational cost vs performance gain")
print("• Report confidence intervals for robust comparisons")
````
